````python
# =============================================================================
# 2. UTILITY FUNCTIONS
# =============================================================================
def parse_file_name(path: Path) -> Tuple[str, int]:
    m = re.match(r"(.+?)_(\d{4})-\d{2}-\d{2}_\d{4}-\d{2}-\d{2}$", path.stem)
    if not m:
        raise ValueError(f"Unexpected file name format: {path.name}")
    return m.group(1), int(m.group(2))


def add_calendar_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    idx = pd.DatetimeIndex(df.index)
    df["hour"] = idx.hour
    df["dayofweek"] = idx.dayofweek
    df["day"] = idx.day
    df["dayofyear"] = idx.dayofyear
    df["month"] = idx.month
    df["year"] = idx.year
    df["season"] = df["month"].map(
        {
            12: "winter",
            1: "winter",
            2: "winter",
            3: "spring",
            4: "spring",
            5: "spring",
            6: "summer",
            7: "summer",
            8: "summer",
            9: "autumn",
            10: "autumn",
            11: "autumn",
        }
    )
    return df


def add_cyclical_features(
    df: pd.DataFrame, has_hour: bool = True
) -> pd.DataFrame:
    df = df.copy()
    if has_hour and "hour" in df.columns:
        df["hour_sin"] = np.sin(2 * np.pi * df["hour"] / 24)
        df["hour_cos"] = np.cos(2 * np.pi * df["hour"] / 24)
    if "month" in df.columns:
        df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
        df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
    if "dayofyear" in df.columns:
        df["doy_sin"] = np.sin(2 * np.pi * df["dayofyear"] / 366)
        df["doy_cos"] = np.cos(2 * np.pi * df["dayofyear"] / 366)
    return df


def safe_mape(y_true, y_pred, eps=1e-6):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return (
        np.mean(np.abs((y_true - y_pred) / np.maximum(np.abs(y_true), eps)))
        * 100
    )


def wape(y_true, y_pred, eps=1e-6):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return (
        np.sum(np.abs(y_true - y_pred))
        / max(np.sum(np.abs(y_true)), eps)
        * 100
    )


def rmse(y_true, y_pred):
    return math.sqrt(mean_squared_error(y_true, y_pred))


def directional_accuracy(y_true_matrix, y_pred_matrix):
    y_true_matrix = np.asarray(y_true_matrix)
    y_pred_matrix = np.asarray(y_pred_matrix)
    if y_true_matrix.ndim == 1 or y_true_matrix.shape[1] < 2:
        return np.nan
    return np.mean(
        np.sign(np.diff(y_true_matrix, axis=1))
        == np.sign(np.diff(y_pred_matrix, axis=1))
    )
````

````python
# =============================================================================
# 3. LOAD DATA
# =============================================================================
files = sorted(DATA_DIR.glob("*.parquet"))
assert files, f"No parquet files found in {DATA_DIR.resolve()}"
frames, inventory = [], []
for path in files:
    city, year = parse_file_name(path)
    df = pd.read_parquet(path)
    if not isinstance(df.index, pd.DatetimeIndex):
        if "time" in df.columns:
            df["time"] = pd.to_datetime(df["time"])
            df = df.set_index("time")
        else:
            raise ValueError(
                f"{path.name}: no DatetimeIndex and no time column"
            )
    missing = sorted(set(FEATURE_COLUMNS) - set(df.columns))
    if missing:
        raise ValueError(f"{path.name}: missing columns {missing}")
    df = df[FEATURE_COLUMNS].sort_index()
    df["city"] = city
    df["year_file"] = year
    df["climate_group"] = CLIMATE_GROUP_MAP.get(city, "unknown")
    frames.append(df)
    inventory.append(
        {
            "file": path.name,
            "city": city,
            "year": year,
            "rows": len(df),
            "start": df.index.min(),
            "end": df.index.max(),
            "missing_values": int(df[FEATURE_COLUMNS].isna().sum().sum()),
            "duplicated_index": int(df.index.duplicated().sum()),
        }
    )
raw_hourly = pd.concat(frames).sort_index()
raw_hourly.index.name = "time"
inventory_df = pd.DataFrame(inventory).sort_values(["city", "year"])
print(
    "Files:",
    len(files),
    "Rows:",
    len(raw_hourly),
    "Cities:",
    sorted(raw_hourly.city.unique()),
)
display(inventory_df)

quality_rows = []
aligned = []
for city, g in raw_hourly.groupby("city"):
    expected = pd.date_range(g.index.min(), g.index.max(), freq="H")
    quality_rows.append(
        {
            "city": city,
            "rows": len(g),
            "expected_hourly_rows": len(expected),
            "missing_timestamps": len(expected.difference(g.index)),
            "duplicated_timestamps": int(g.index.duplicated().sum()),
            "missing_values": int(g[FEATURE_COLUMNS].isna().sum().sum()),
        }
    )
    vals = (
        g[FEATURE_COLUMNS]
        .resample("H")
        .mean(numeric_only=True)
        .interpolate(method="time", limit_direction="both")
    )
    vals["city"] = city
    vals["climate_group"] = CLIMATE_GROUP_MAP.get(city, "unknown")
    vals["year_file"] = vals.index.year
    aligned.append(vals)
quality_df = pd.DataFrame(quality_rows)
raw_hourly = pd.concat(aligned).sort_index()
display(quality_df)
````

````python
# Daily aggregation for EDA and modeling
parts = []
for city, g in raw_hourly.groupby("city"):
    d = g[FEATURE_COLUMNS].resample("D").agg(DAILY_AGG)
    d["city"] = city
    d["climate_group"] = CLIMATE_GROUP_MAP.get(city, "unknown")
    parts.append(d)
daily = pd.concat(parts).sort_index()
daily.index.name = "date"
daily = add_cyclical_features(add_calendar_columns(daily), has_hour=False)
print("Daily shape:", daily.shape)
display(daily.head())
display(
    daily.groupby("city")[FEATURE_COLUMNS]
    .agg(["mean", "std", "min", "max"])
    .round(3)
)
````

````python
# =============================================================================
# 4. FEATURE ENGINEERING
# =============================================================================
def make_daily_features(daily_df: pd.DataFrame) -> pd.DataFrame:
    parts = []
    for city, g in daily_df.groupby("city"):
        g = add_cyclical_features(
            add_calendar_columns(g.sort_index().copy()), has_hour=False
        )
        g["temp_diff_1"] = g["temperature_2m"].diff()
        g["temp_accel_1"] = g["temp_diff_1"].diff()
        g["pressure_diff_1"] = g["surface_pressure"].diff()
        for col in [
            "temperature_2m",
            "relative_humidity_2m",
            "precipitation",
            "wind_speed_10m",
            "surface_pressure",
        ]:
            for lag in [1, 2, 3, 7, 14, 30]:
                g[f"{col}_lag_{lag}"] = g[col].shift(lag)
            for win in [7, 14, 30]:
                r = g[col].rolling(win, min_periods=max(3, win // 3))
                g[f"{col}_roll_mean_{win}"] = r.mean()
                g[f"{col}_roll_std_{win}"] = r.std()
                if col in ["temperature_2m", "surface_pressure"]:
                    g[f"{col}_roll_min_{win}"] = r.min()
                    g[f"{col}_roll_max_{win}"] = r.max()
        for col in ["temperature_2m", "precipitation", "surface_pressure"]:
            norm = g.groupby("dayofyear")[col].transform("mean")
            g[f"{col}_climate_norm_doy"] = norm
            g[f"{col}_anomaly_vs_norm"] = g[col] - norm
        parts.append(g)
    return pd.concat(parts).sort_index()


features_daily = make_daily_features(daily)
print("Feature matrix:", features_daily.shape)
display(features_daily.head())
````

````python
def make_classification_windows(
    feature_df,
    window_days=CLASSIFICATION_WINDOW_DAYS,
    step_days=CLASSIFICATION_STEP_DAYS,
    target=CLASSIFICATION_TARGET,
):
    rows = []
    base_cols = FEATURE_COLUMNS + ["temp_diff_1", "pressure_diff_1"]
    for city, g in feature_df.groupby("city"):
        g = g.sort_index().copy()
        label = (
            city
            if target == "city"
            else CLIMATE_GROUP_MAP.get(city, "unknown")
        )
        for start in range(0, len(g) - window_days + 1, step_days):
            w = g.iloc[start : start + window_days]
            row = {
                "city": city,
                "climate_group": CLIMATE_GROUP_MAP.get(city, "unknown"),
                "target": label,
                "window_start": w.index.min(),
                "window_end": w.index.max(),
                "year": int(w.index.max().year),
                "month": int(w.index.max().month),
                "season": w["season"].mode().iloc[0],
            }
            for col in base_cols:
                s = w[col].dropna()
                row[f"{col}_mean"] = s.mean()
                row[f"{col}_std"] = s.std()
                row[f"{col}_min"] = s.min()
                row[f"{col}_max"] = s.max()
                row[f"{col}_median"] = s.median()
                row[f"{col}_q10"] = s.quantile(0.1)
                row[f"{col}_q90"] = s.quantile(0.9)
                if col in ["temperature_2m", "surface_pressure"]:
                    row[f"{col}_amplitude"] = s.max() - s.min()
            row["precipitation_days"] = int((w["precipitation"] > 0).sum())
            row["snowfall_days"] = int((w["snowfall"] > 0).sum())
            row["max_wind"] = w["wind_speed_10m"].max()
            row["mean_temp"] = w["temperature_2m"].mean()
            row["std_temp"] = w["temperature_2m"].std()
            rows.append(row)
    return pd.DataFrame(rows)


classification_windows = make_classification_windows(features_daily)
print("Classification windows:", classification_windows.shape)
display(classification_windows.head())
display(classification_windows["target"].value_counts())
````

</h2>

<div style="color: #ffffff; line-height: 1.8; font-size: 1.02em;">
Задача классификации решает маршрутизацию: по окну метеопараметров модель определяет город или климатическую группу. Разбиение строго временное: 2019-2023 для обучения, 2024 для validation и 2025 для финального test.
</div>

<div style="background: #251a3f; border-left: 4px solid #b794f4; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #ffffff;">
  <div style="font-weight: 700; color: #ffffff; margin-bottom: 6px; font-size: 0.95em;">📐 Ключевая формула</div>
  <div style="line-height: 1.7;">Основная метрика выбора модели — macro-F1:<br>
  
$$F1_{macro}=\frac{1}{K}\sum_{k=1}^{K}\frac{2\cdot Precision_k\cdot Recall_k}{Precision_k+Recall_k}$$

Где: 
- $K$ — число классов; 
- $Precision_k$ — точность для класса $k$; 
- $Recall_k$ — полнота для класса $k$. 

Метрика равноправно учитывает каждый город, поэтому слабое качество на одном классе не маскируется общей долей правильных ответов.
</div>

<div style="background: #1a202c; color: #ffffff; padding: 8px 16px; border-radius: 10px 10px 0 0; margin-bottom: -15px; font-family: 'Fira Code', monospace; font-size: 0.8em; display: flex; justify-content: space-between; align-items: center;">
  <span>🐍 classification_training.py</span>
  <span style="background: #2d3748; padding: 2px 10px; border-radius: 10px; font-size: 0.85em;">Python</span>
</div>

````python
# =============================================================================
# 5. CLASSIFICATION
# =============================================================================
train_w = classification_windows[
    classification_windows.year <= TRAIN_END_YEAR
].copy()
val_w = classification_windows[classification_windows.year == VAL_YEAR].copy()
test_w = classification_windows[
    classification_windows.year == TEST_YEAR
].copy()
print({"train": len(train_w), "val": len(val_w), "test": len(test_w)})
median_values = (
    train_w[clf_feature_cols]
    .replace([np.inf, -np.inf], np.nan)
    .median(numeric_only=True)
)


def xy_clf(df):
    return (
        df[clf_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(median_values),
        df["target"],
    )


X_train, y_train = xy_clf(train_w)
X_val, y_val = xy_clf(val_w)
X_test, y_test = xy_clf(test_w)
models = {
    "logistic_regression": Pipeline(
        [
            ("scaler", StandardScaler()),
            (
                "model",
                LogisticRegression(
                    max_iter=2000,
                    class_weight="balanced",
                    random_state=RANDOM_STATE,
                ),
            ),
        ]
    ),
    "random_forest": RandomForestClassifier(
        n_estimators=N_ESTIMATORS_CLASSIFICATION,
        min_samples_leaf=2,
        class_weight="balanced_subsample",
        random_state=RANDOM_STATE,
        n_jobs=N_JOBS,
    ),
    "extra_trees": ExtraTreesClassifier(
        n_estimators=N_ESTIMATORS_CLASSIFICATION,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=N_JOBS,
    ),
}
trained_classifiers = {}
rows = []
for name, model in models.items():
    model.fit(X_train, y_train)
    trained_classifiers[name] = model
    pv = model.predict(X_val)
    pt = model.predict(X_test)
    rows.append(
        {
            "model": name,
            "val_accuracy": accuracy_score(y_val, pv),
            "val_f1_macro": f1_score(y_val, pv, average="macro"),
            "test_accuracy": accuracy_score(y_test, pt),
            "test_f1_macro": f1_score(y_test, pt, average="macro"),
            "test_f1_weighted": f1_score(y_test, pt, average="weighted"),
        }
    )
clf_results_df = pd.DataFrame(rows).sort_values(
    "val_f1_macro", ascending=False
)
display(clf_results_df)
best_clf_name = clf_results_df.iloc[0].model
best_classifier = trained_classifiers[best_clf_name]
print("Best classifier:", best_clf_name)
````

</h2>

<div style="color: #ffffff; line-height: 1.8; font-size: 1.02em;">
Для каждого города строится отдельная supervised-задача: признаки описывают последние 90 дней и календарный контекст, а цель содержит 30 будущих среднесуточных температур.
</div>

<div style="background: #251a3f; border-left: 4px solid #b794f4; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #ffffff;">
  <div style="font-weight: 700; color: #ffffff; margin-bottom: 6px; font-size: 0.95em;">📐 Ключевая формула</div>

  <div style="line-height: 1.7;">Прогноз строится как multi-output задача:<br>
  
$$\hat{y}_{t+1:t+30}=f(x_t)$$

Где: 

- $x_t$ — признаки, доступные на дату прогноза $t$; 
- $\hat{y}_{t+1:t+30}$ — вектор прогнозов среднесуточной температуры на 30 будущих горизонтов; 
- $f$ — городская модель, выбранная после маршрутизации.
</div>

<div style="background: #1a202c; color: #ffffff; padding: 8px 16px; border-radius: 10px 10px 0 0; margin-bottom: -15px; font-family: 'Fira Code', monospace; font-size: 0.8em; display: flex; justify-content: space-between; align-items: center;">

  <span style="background: #2d3748; padding: 2px 10px; border-radius: 10px; font-size: 0.85em;">Python</span>

````python
# =============================================================================
# 7. FORECASTING
# =============================================================================
def train_temperature_climatology(g: pd.DataFrame) -> Tuple[pd.Series, float]:
    """Return day-of-year temperature norm computed only from train years.

    This avoids target/test leakage in future_climate_norm_temp_h* features and
    in the climate_norm baseline used on validation/test periods.
    """
    train_hist = g[g.index.year <= TRAIN_END_YEAR].copy()
    if train_hist.empty:
        train_hist = g.copy()
    by_doy = train_hist.groupby("dayofyear")["temperature_2m"].mean()
    fallback = float(train_hist["temperature_2m"].mean())
    return by_doy, fallback


def climatology_value(
    norm_by_doy: pd.Series, fallback: float, doy: int
) -> float:
    value = norm_by_doy.get(int(doy), np.nan)
    return float(fallback if pd.isna(value) else value)


def make_forecast_supervised(
    feature_df,
    lookback_days=FORECAST_LOOKBACK_DAYS,
    horizon_days=FORECAST_HORIZON_DAYS,
):
    datasets = {}
    all_rows = []
    for city, g in feature_df.groupby("city"):
        g = g.sort_index().copy()
        rows = []
        temp_norm_by_doy, temp_norm_fallback = train_temperature_climatology(g)
        for i in range(lookback_days, len(g) - horizon_days):
            hist = g.iloc[i - lookback_days : i]
            cur = g.iloc[i]
            future = (
                g["temperature_2m"].iloc[i + 1 : i + horizon_days + 1].values
            )
            row = {
                "city": city,
                "climate_group": CLIMATE_GROUP_MAP.get(city, "unknown"),
                "as_of_date": g.index[i],
                "year": int(g.index[i].year),
                "month": int(g.index[i].month),
                "dayofyear": int(g.index[i].dayofyear),
            }
            for col in [
                "temperature_2m",
                "relative_humidity_2m",
                "precipitation",
                "rain",
                "snowfall",
                "weathercode",
                "wind_speed_10m",
                "surface_pressure",
                "temp_diff_1",
                "temp_accel_1",
                "pressure_diff_1",
                "month_sin",
                "month_cos",
                "doy_sin",
                "doy_cos",
            ]:
                if col in g.columns:
                    row[f"current_{col}"] = cur[col]
            for col in [
                "temperature_2m",
                "relative_humidity_2m",
                "precipitation",
                "wind_speed_10m",
                "surface_pressure",
            ]:
                for win in [7, 14, 30, 60, 90]:
                    s = hist[col].tail(win)
                    row[f"hist_{col}_mean_{win}"] = s.mean()
                    row[f"hist_{col}_std_{win}"] = s.std()
                    row[f"hist_{col}_min_{win}"] = s.min()
                    row[f"hist_{col}_max_{win}"] = s.max()
                    if col in ["temperature_2m", "surface_pressure"]:
                        row[f"hist_{col}_trend_{win}"] = s.iloc[-1] - s.iloc[0]
            row["hist_precip_days_30"] = int(
                (hist["precipitation"].tail(30) > 0).sum()
            )
            row["hist_snow_days_30"] = int(
                (hist["snowfall"].tail(30) > 0).sum()
            )
            for h in range(1, horizon_days + 1):
                target_date = g.index[i] + pd.Timedelta(days=h)
                doy = target_date.dayofyear
                row[f"future_climate_norm_temp_h{h}"] = climatology_value(
                    temp_norm_by_doy, temp_norm_fallback, doy
                )
                row[f"y_h{h}"] = future[h - 1]
            rows.append(row)
        city_df = pd.DataFrame(rows)
        datasets[city] = city_df
        all_rows.append(city_df)
    return pd.concat(all_rows, ignore_index=True), datasets


forecast_df, forecast_by_city = make_forecast_supervised(features_daily)
print("Forecast supervised:", forecast_df.shape)
display(forecast_df.head())
````

````python
y_cols = [f"y_h{h}" for h in range(1, FORECAST_HORIZON_DAYS + 1)]
forecast_feature_cols = [
    c
    for c in forecast_df.columns
    if c not in ["city", "climate_group", "as_of_date"] + y_cols
]
forecast_models = {}
forecast_results = []
forecast_predictions = []
for city, city_df in tqdm(
    forecast_by_city.items(), desc="Training forecast models"
):
    train_df = city_df[city_df.year <= TRAIN_END_YEAR].copy()
    val_df = city_df[city_df.year == VAL_YEAR].copy()
    test_df = city_df[city_df.year == TEST_YEAR].copy()
    med = (
        train_df[forecast_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .median(numeric_only=True)
    )
    Xtr = (
        train_df[forecast_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(med)
    )
    Ytr = train_df[y_cols]
    Xval = (
        val_df[forecast_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(med)
    )
    Yval = val_df[y_cols]
    Xte = (
        test_df[forecast_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(med)
    )
    Yte = test_df[y_cols]
    candidates = {
        "random_forest": RandomForestRegressor(
            n_estimators=N_ESTIMATORS_FORECAST,
            min_samples_leaf=2,
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
        ),
        "extra_trees": ExtraTreesRegressor(
            n_estimators=N_ESTIMATORS_FORECAST,
            min_samples_leaf=2,
            random_state=RANDOM_STATE,
            n_jobs=N_JOBS,
        ),
        "ridge_baseline": Pipeline(
            [("scaler", StandardScaler()), ("model", Ridge(alpha=5.0))]
        ),
    }
    fitted = {}
    scores = []
    for name, model in candidates.items():
        model.fit(Xtr, Ytr)
        fitted[name] = model
        pe = model.predict(Xval if len(Xval) else Xte)
        ye = Yval if len(Xval) else Yte
        scores.append(
            {"model": name, "val_or_test_mae": mean_absolute_error(ye, pe)}
        )
    best_name = (
        pd.DataFrame(scores).sort_values("val_or_test_mae").iloc[0].model
    )
    best_model = fitted[best_name]
    pred = best_model.predict(Xte)
    forecast_models[city] = {
        "model_name": best_name,
        "model": best_model,
        "median_values": med,
        "feature_cols": forecast_feature_cols,
        "target_cols": y_cols,
    }
    forecast_results.append(
        {
            "city": city,
            "model": best_name,
            "test_MAE_all_horizons": mean_absolute_error(Yte, pred),
            "test_RMSE_all_horizons": rmse(Yte.values.ravel(), pred.ravel()),
            "test_MAPE_all_horizons": safe_mape(
                Yte.values.ravel(), pred.ravel()
            ),
            "test_WAPE_all_horizons": wape(Yte.values.ravel(), pred.ravel()),
            "directional_accuracy": directional_accuracy(Yte.values, pred),
            "train_rows": len(train_df),
            "val_rows": len(val_df),
            "test_rows": len(test_df),
        }
    )
    pdf = test_df[["city", "as_of_date", "year", "month"]].copy()
    for j, col in enumerate(y_cols):
        pdf[f"true_{col}"] = Yte[col].values
        pdf[f"pred_{col}"] = pred[:, j]
        pdf[f"err_{col}"] = pdf[f"true_{col}"] - pdf[f"pred_{col}"]
    forecast_predictions.append(pdf)
forecast_results_df = pd.DataFrame(forecast_results).sort_values(
    "test_MAE_all_horizons"
)
forecast_pred_df = pd.concat(forecast_predictions, ignore_index=True)
display(forecast_results_df)
````

````python
# =============================================================================
# FORECAST BASELINES: persistence and climate norm
# =============================================================================
baseline_rows = []
for city, city_df in forecast_by_city.items():
    test_df = city_df[city_df.year == TEST_YEAR].copy()
    if test_df.empty:
        continue
    Y_true = test_df[y_cols].values
    # Persistence: use current temperature for all future horizons.
    persistence_pred = np.repeat(
        test_df[["current_temperature_2m"]].values,
        FORECAST_HORIZON_DAYS,
        axis=1,
    )
    # Climate norm: use precomputed future climate norm columns.
    norm_cols = [
        f"future_climate_norm_temp_h{h}"
        for h in range(1, FORECAST_HORIZON_DAYS + 1)
    ]
    climate_norm_pred = test_df[norm_cols].values
    for baseline_name, pred in [
        ("persistence", persistence_pred),
        ("climate_norm", climate_norm_pred),
    ]:
        baseline_rows.append(
            {
                "city": city,
                "model": baseline_name,
                "test_MAE_all_horizons": mean_absolute_error(Y_true, pred),
                "test_RMSE_all_horizons": rmse(Y_true.ravel(), pred.ravel()),
                "test_MAPE_all_horizons": safe_mape(
                    Y_true.ravel(), pred.ravel()
                ),
                "test_WAPE_all_horizons": wape(Y_true.ravel(), pred.ravel()),
                "directional_accuracy": directional_accuracy(Y_true, pred),
            }
        )
forecast_baselines_df = pd.DataFrame(baseline_rows)
model_vs_baseline_df = pd.concat(
    [
        forecast_results_df[
            [
                "city",
                "model",
                "test_MAE_all_horizons",
                "test_RMSE_all_horizons",
                "test_MAPE_all_horizons",
                "test_WAPE_all_horizons",
                "directional_accuracy",
            ]
        ],
        forecast_baselines_df,
    ],
    ignore_index=True,
).sort_values(["city", "test_MAE_all_horizons"])
display(model_vs_baseline_df)

# Compact improvement table: best ML model vs best baseline by city.
improvement_rows = []
for city in forecast_results_df.city.unique():
    ml = forecast_results_df[forecast_results_df.city == city].iloc[0]
    base = (
        forecast_baselines_df[forecast_baselines_df.city == city]
        .sort_values("test_MAE_all_horizons")
        .iloc[0]
    )
    improvement_rows.append(
        {
            "city": city,
            "ml_model": ml["model"],
            "ml_MAE": ml["test_MAE_all_horizons"],
            "best_baseline": base["model"],
            "baseline_MAE": base["test_MAE_all_horizons"],
            "MAE_improvement_vs_baseline": base["test_MAE_all_horizons"]
            - ml["test_MAE_all_horizons"],
            "relative_MAE_improvement_pct": (
                base["test_MAE_all_horizons"] - ml["test_MAE_all_horizons"]
            )
            / base["test_MAE_all_horizons"]
            * 100,
        }
    )
forecast_improvement_df = pd.DataFrame(improvement_rows).sort_values(
    "relative_MAE_improvement_pct", ascending=False
)
display(forecast_improvement_df)
````

````python
# Residual diagnostics for h=1
# Shapiro-Wilk доступен через scipy и выполняется всегда.
# Ljung-Box и White test требуют statsmodels; если пакет не установлен, выводится явное предупреждение.
rows = []
for city in forecast_pred_df.city.unique():
    sub = forecast_pred_df[forecast_pred_df.city == city].sort_values(
        "as_of_date"
    )
    residuals = sub.err_y_h1.dropna().values
    if len(residuals) > 8:
        sh_stat, sh_p = shapiro(residuals[: min(len(residuals), 5000)])
        row = {
            "city": city,
            "shapiro_statistic": float(sh_stat),
            "shapiro_p_normality": float(sh_p),
            "residual_mean_h1": float(np.mean(residuals)),
            "residual_std_h1": float(np.std(residuals)),
        }
        if STATSMODELS_AVAILABLE:
            lb = acorr_ljungbox(residuals, lags=[10], return_df=True)
            fitted = sub.pred_y_h1.dropna().values
            exog = sm.add_constant(
                np.column_stack([fitted, np.arange(len(fitted))])
            )
            try:
                white_stat, white_p, _, _ = het_white(residuals, exog)
            except Exception:
                white_stat, white_p = np.nan, np.nan
            row.update(
                {
                    "ljungbox_p_autocorr_lag10": float(lb.lb_pvalue.iloc[0]),
                    "white_p_heteroskedasticity": (
                        float(white_p) if pd.notna(white_p) else np.nan
                    ),
                }
            )
        else:
            row.update(
                {
                    "ljungbox_p_autocorr_lag10": np.nan,
                    "white_p_heteroskedasticity": np.nan,
                }
            )
        rows.append(row)
residual_diag_df = pd.DataFrame(rows)
display(residual_diag_df)
if not STATSMODELS_AVAILABLE:
    print(
        "Выполнен только Shapiro-Wilk. Для Ljung-Box и White test установите statsmodels и перезапустите notebook: pip install statsmodels"
    )
````

````python
# Forecast next 30 days after the last available observation
# Unlike the supervised test rows, this block builds feature rows with as_of_date = last available date.
def build_future_forecast_row(
    city: str, as_of_date: Optional[pd.Timestamp] = None
) -> pd.DataFrame:
    g = features_daily[features_daily.city == city].sort_index().copy()
    if g.empty:
        raise ValueError(f"No data for {city}")
    if as_of_date is None:
        as_of_date = g.index.max()
    as_of_date = pd.Timestamp(as_of_date)
    g_hist = g[g.index <= as_of_date]
    if len(g_hist) < FORECAST_LOOKBACK_DAYS:
        raise ValueError(f"Not enough history for {city}: {len(g_hist)} rows")
    hist = g_hist.tail(FORECAST_LOOKBACK_DAYS)
    cur = hist.iloc[-1]
    temp_norm_by_doy, temp_norm_fallback = train_temperature_climatology(g)
    row = {
        "city": city,
        "climate_group": CLIMATE_GROUP_MAP.get(city, "unknown"),
        "as_of_date": as_of_date,
        "year": int(as_of_date.year),
        "month": int(as_of_date.month),
        "dayofyear": int(as_of_date.dayofyear),
    }
    for col in [
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "rain",
        "snowfall",
        "weathercode",
        "wind_speed_10m",
        "surface_pressure",
        "temp_diff_1",
        "temp_accel_1",
        "pressure_diff_1",
        "month_sin",
        "month_cos",
        "doy_sin",
        "doy_cos",
    ]:
        if col in g.columns:
            row[f"current_{col}"] = cur[col]
    for col in [
        "temperature_2m",
        "relative_humidity_2m",
        "precipitation",
        "wind_speed_10m",
        "surface_pressure",
    ]:
        for win in [7, 14, 30, 60, 90]:
            s = hist[col].tail(win)
            row[f"hist_{col}_mean_{win}"] = s.mean()
            row[f"hist_{col}_std_{win}"] = s.std()
            row[f"hist_{col}_min_{win}"] = s.min()
            row[f"hist_{col}_max_{win}"] = s.max()
            if col in ["temperature_2m", "surface_pressure"]:
                row[f"hist_{col}_trend_{win}"] = s.iloc[-1] - s.iloc[0]
    row["hist_precip_days_30"] = int(
        (hist["precipitation"].tail(30) > 0).sum()
    )
    row["hist_snow_days_30"] = int((hist["snowfall"].tail(30) > 0).sum())
    for h in range(1, FORECAST_HORIZON_DAYS + 1):
        target_date = as_of_date + pd.Timedelta(days=h)
        row[f"future_climate_norm_temp_h{h}"] = climatology_value(
            temp_norm_by_doy, temp_norm_fallback, target_date.dayofyear
        )
    return pd.DataFrame([row])


future_forecasts = []
for city, pack in forecast_models.items():
    row = build_future_forecast_row(city)
    X = (
        row[pack["feature_cols"]]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(pack["median_values"])
    )
    pred = pack["model"].predict(X)[0]
    as_of = row.as_of_date.iloc[0]
    for h, v in enumerate(pred, start=1):
        future_forecasts.append(
            {
                "city": city,
                "as_of_date": as_of,
                "forecast_date": as_of + pd.Timedelta(days=h),
                "horizon_day": h,
                "forecast_temperature_2m": v,
            }
        )
future_forecast_df = pd.DataFrame(future_forecasts)
display(future_forecast_df.head(40))
plt.figure(figsize=(14, 6))
sns.lineplot(
    data=future_forecast_df,
    x="forecast_date",
    y="forecast_temperature_2m",
    hue="city",
    marker="o",
)
plt.xticks(rotation=30)
plt.title("Прогноз на следующий месяц после последней доступной даты")
plt.show()
````

    Классификатор работает как диспетчер: он определяет, какой климатический профиль больше похож на последнее окно наблюдений. Затем forecast-модель, обученная на соответствующем городе, строит численный прогноз. Разделение снижает риск смешать разные климатические режимы в одной усредненной модели.

````python
# =============================================================================
# 9. INTEGRATED PIPELINE
# =============================================================================
def build_single_classification_window(city, end_date=None):
    g = features_daily[features_daily.city == city].sort_index().copy()
    if end_date is not None:
        g = g[g.index <= pd.Timestamp(end_date)]
    w = g.tail(CLASSIFICATION_WINDOW_DAYS)
    if len(w) < CLASSIFICATION_WINDOW_DAYS:
        raise ValueError(f"Not enough data for {city}")
    tmp = make_classification_windows(
        w,
        window_days=CLASSIFICATION_WINDOW_DAYS,
        step_days=CLASSIFICATION_WINDOW_DAYS,
        target=CLASSIFICATION_TARGET,
    )
    return (
        tmp[clf_feature_cols]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(median_values)
    )


def forecast_for_city(city):
    pack = forecast_models[city]
    row = build_future_forecast_row(city)
    X = (
        row[pack["feature_cols"]]
        .replace([np.inf, -np.inf], np.nan)
        .fillna(pack["median_values"])
    )
    pred = pack["model"].predict(X)[0]
    as_of = row.as_of_date.iloc[0]
    return pd.DataFrame(
        {
            "city": city,
            "as_of_date": as_of,
            "forecast_date": [
                as_of + pd.Timedelta(days=h)
                for h in range(1, FORECAST_HORIZON_DAYS + 1)
            ],
            "horizon_day": range(1, FORECAST_HORIZON_DAYS + 1),
            "forecast_temperature_2m": pred,
        }
    )


def two_stage_predict(city_for_input_window):
    X = build_single_classification_window(city_for_input_window)
    predicted_route = best_classifier.predict(X)[0]
    proba_map = {}
    if hasattr(best_classifier, "predict_proba"):
        proba_map = dict(
            zip(best_classifier.classes_, best_classifier.predict_proba(X)[0])
        )
    route_city = (
        predicted_route
        if CLASSIFICATION_TARGET == "city"
        else city_for_input_window
    )
    return {
        "input_city_for_demo": city_for_input_window,
        "classification_target": CLASSIFICATION_TARGET,
        "predicted_route": predicted_route,
        "route_city_for_forecast_model": route_city,
        "classification_probabilities": proba_map,
        "forecast": forecast_for_city(route_city),
    }


pipeline_demo = []
for city in sorted(features_daily.city.unique()):
    r = two_stage_predict(city)
    f = r["forecast"]
    pipeline_demo.append(
        {
            "input_city": city,
            "predicted_route": r["predicted_route"],
            "route_city_for_forecast_model": r[
                "route_city_for_forecast_model"
            ],
            "first_forecast_date": f.forecast_date.min(),
            "last_forecast_date": f.forecast_date.max(),
            "mean_forecast_temp": f.forecast_temperature_2m.mean(),
        }
    )
pipeline_demo_df = pd.DataFrame(pipeline_demo)
display(pipeline_demo_df)
````

### Финальная интерпретация после запуска

#### 1. Качество данных

Данные загружены из 42 parquet-файлов: 6 городов × 7 лет. Временная сетка регулярная: для каждого города ожидаемые и фактические hourly-строки совпадают, пропущенных timestamps нет. Это позволяет использовать честное временное разбиение train/validation/test без сложной предварительной импутации.

#### 2. EDA и климатические различия

EDA подтверждает выраженную годовую сезонность температуры и заметные различия между городами. Благовещенск имеет наиболее континентальный профиль с очень холодной зимой и большой годовой амплитудой. Сочи и Геленджик формируют теплый черноморский профиль. Москва и Санкт-Петербург ближе по среднегодовому уровню, а значит их сложнее устойчиво различать только по агрегированным окнам.

Осадки, дождь и снег имеют негауссовы распределения с большим числом нулей. IQR-выбросы по этим признакам не удаляются автоматически, потому что они могут соответствовать реальным экстремальным погодным событиям.

#### 3. Стационарность

В текущем запуске `statsmodels` отсутствует, поэтому ADF-тест не был выполнен. Это не блокирует используемые ML-модели, потому что нестационарность частично учитывается календарными, циклическими, лаговыми, rolling-признаками и климатической нормой. Для полного статистического отчета нужно установить `statsmodels` и перезапустить notebook.

#### 4. Классификация

Основная модель классификации выбирается по validation macro-F1. По этому критерию выбрана Extra Trees (`val_f1_macro = 0.9677`). На test она дает `accuracy = 0.8113` и `macro-F1 = 0.7932`. Logistic Regression на test показывает немного больший macro-F1 (`0.7955`), но использовать test для выбора модели нельзя: test предназначен только для финальной независимой оценки.

Confusion matrix показывает, что главный проблемный класс — Москва: recall `0.2830`, большинство ошибок Москвы уходят в Благовещенск. Это может быть следствием межгодового сдвига, сезонного сходства отдельных окон или переопоры модели на признаки давления и ветра.

Сезонная диагностика логична: зимой ошибка минимальна (`0.0714`), летом максимальна (`0.3205`). Зимой климатические различия городов выражены сильнее, а летом температурные профили частично сближаются.

#### 5. Feature importance и robustness

Top features в `importance_df` преимущественно связаны с `surface_pressure` и `wind_speed_10m`. Ablation без pressure-признаков снижает test macro-F1 с `0.7932` до `0.6128`. Это честное ограничение модели: для city-routing давление информативно, но для абстрактной классификации климата такая зависимость может быть нежелательной.

Классификация климатических групп проще: `climate_group` дает `test_accuracy = 0.8585` и `test_f1_macro = 0.8495`. Однако группа `temperate_continental_urban` имеет recall `0.5849`, то есть часть окон Москвы/Санкт-Петербурга всё еще путается с дальневосточным континентально-муссонным профилем.

#### 6. Прогнозирование

Для каждого города обучена отдельная модель 30-дневного прогноза температуры. По MAE качество убывает так: Находка (`2.2067`), Сочи (`2.5281`), Благовещенск (`2.8348`), Геленджик (`3.0465`), Санкт-Петербург (`3.5589`), Москва (`4.1384`).

Сравнивать города лучше по MAE, RMSE и WAPE. MAPE для температуры в градусах Цельсия интерпретируется ограниченно: температура может быть отрицательной или близкой к нулю, поэтому процентная ошибка становится искусственно огромной.

Directional accuracy около `0.51–0.56` означает, что модели в основном воспроизводят сезонный уровень температуры, но слабо угадывают направление краткосрочных изменений. Это ожидаемо для month-ahead прогноза без внешних синоптических/NWP-признаков.

#### 7. Baseline comparison

`climate_norm`, рассчитанный только по train-периоду, является сильным baseline. ML-модели лучше него для Благовещенска, Находки и Сочи, но хуже для Геленджика, Санкт-Петербурга и Москвы. Поэтому общий вывод должен быть осторожным: pipeline работает, но прогнозная часть пока не дает стабильного выигрыша над сезонной нормой для всех городов.

#### 8. Анализ остатков

В текущем запуске выполнен только Shapiro-Wilk: `statsmodels` недоступен, поэтому Ljung-Box и White test не выполнялись, а их p-value равны `NaN`. Нормальность остатков отвергается для 5 из 6 городов; исключение — Геленджик, где `p ≈ 0.076`. Это не ломает ML-подход, но показывает, что распределение ошибок не идеально гауссово.
