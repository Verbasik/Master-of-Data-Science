```python
# ── Стандартная библиотека ──
from __future__ import annotations
from pathlib import Path
from time import perf_counter
from typing import Any, Dict, Iterable, List, Optional, Tuple

# ── Сторонние библиотеки: данные ──
import numpy as np
import pandas as pd

# ── Сторонние библиотеки: визуализация ──
import matplotlib.pyplot as plt
import seaborn as sns

# ── Сторонние библиотеки: статистика ──
from scipy import stats
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.tools.tools import add_constant
import statsmodels.api as sm

# ── Сторонние библиотеки: ML ──
try:
    import shap
    SHAP_AVAILABLE = True
except ImportError:
    SHAP_AVAILABLE = False
    print("SHAP не установлен. pip install shap")
    
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Настраиваем базовый стиль графиков для всего ноутбука
sns.set_theme(style="whitegrid", context="talk")

# Создаём базовый путь к каталогу с данными
DATA_DIR = Path("/content")

# Фиксируем список тестируемых лабораторных задержек в минутах
LAB_DELAY_CANDIDATES = list(range(10, 16))

# Определяем набор окон для скользящих статистик
ROLLING_WINDOWS = [5, 15, 30, 60]

# Определяем набор лагов для инженерии признаков
LAG_STEPS = [1, 5, 10, 15, 30, 60]

# Ограничиваем количество финальных признаков для OLS-модели
OLS_FEATURE_LIMIT = 20

# Включаем более аккуратный вывод чисел в pandas
pd.set_option("display.float_format", lambda value: f"{value:0.4f}")


# Безопасный расчёт MAPE с защитой от деления на ноль
def safe_mape(y_true: pd.Series, y_pred: pd.Series) -> float:
    """
    Description:
    ---------------
        Рассчитывает MAPE только по тем наблюдениям, где истинное
        значение отлично от нуля, чтобы избежать деления на ноль.

    Args:
    ---------------
        y_true: Фактические значения целевой переменной
        y_pred: Предсказанные значения модели

    Returns:
    ---------------
        Значение MAPE в процентах

    Raises:
    ---------------
        ValueError: Если после фильтрации не осталось валидных строк

    Examples:
    ---------------
        >>> safe_mape(pd.Series([1.0, 2.0]), pd.Series([1.1, 1.9]))
        7.5
    """
    # Создаём булеву маску только для ненулевых фактических значений
    non_zero_mask = y_true != 0

    # Проверяем, что после фильтрации остались наблюдения
    if non_zero_mask.sum() == 0:
        raise ValueError("Невозможно вычислить MAPE: все y_true равны нулю.")

    # Вычисляем относительную ошибку только на валидных строках
    relative_error = (
        (y_true.loc[non_zero_mask] - y_pred.loc[non_zero_mask]).abs()
        / y_true.loc[non_zero_mask].abs()
    )

    # Возвращаем среднюю относительную ошибку в процентах
    return float(relative_error.mean() * 100)


# Расчёт WAPE как более устойчивой бизнес-метрики для производства
def safe_wape(y_true: pd.Series, y_pred: pd.Series) -> float:
    """
    Description:
    ---------------
        Рассчитывает Weighted Absolute Percentage Error как отношение
        суммы абсолютных ошибок к сумме абсолютных фактических значений.

    Args:
    ---------------
        y_true: Фактические значения целевой переменной
        y_pred: Предсказанные значения модели

    Returns:
    ---------------
        Значение WAPE в процентах

    Raises:
    ---------------
        ValueError: Если знаменатель равен нулю

    Examples:
    ---------------
        >>> safe_wape(pd.Series([1.0, 2.0]), pd.Series([1.1, 1.9]))
        6.666666666666667
    """
    # Вычисляем знаменатель формулы WAPE
    denominator = y_true.abs().sum()

    # Проверяем корректность знаменателя
    if denominator == 0:
        raise ValueError("Невозможно вычислить WAPE: сумма |y_true| равна нулю.")

    # Вычисляем сумму абсолютных ошибок и переводим в проценты
    return float((y_true.sub(y_pred).abs().sum() / denominator) * 100)


# Расчёт метрик направления изменения процесса во времени
def compute_directional_metrics(
    y_true: pd.Series,
    y_pred: pd.Series,
) -> Dict[str, float]:
    """
    Description:
    ---------------
        Считает метрики, которые оценивают не только уровень ошибки,
        но и способность модели угадывать направление изменения цели.

    Args:
    ---------------
        y_true: Фактические значения целевой переменной
        y_pred: Предсказанные значения модели

    Returns:
    ---------------
        Словарь с directional accuracy и mean signed error

    Raises:
    ---------------
        ValueError: Если в рядах меньше двух наблюдений

    Examples:
    ---------------
        >>> compute_directional_metrics(
        ...     pd.Series([1.0, 2.0, 1.5]),
        ...     pd.Series([1.1, 1.9, 1.6])
        ... )
        {'directional_accuracy': 1.0, 'mean_signed_error': 0.03333333333333336}
    """
    # Проверяем, что ряд достаточно длинный для расчёта приращений
    if len(y_true) < 2 or len(y_pred) < 2:
        raise ValueError("Для directional-метрик требуется минимум 2 точки.")

    # Вычисляем фактические приращения целевой переменной
    true_delta = y_true.diff().dropna()

    # Вычисляем предсказанные приращения целевой переменной
    pred_delta = y_pred.diff().dropna()

    # Синхронизируем индексы после операции diff
    aligned_index = true_delta.index.intersection(pred_delta.index)

    # Сравниваем знак фактических и предсказанных изменений
    directional_accuracy = (
        np.sign(true_delta.loc[aligned_index])
        == np.sign(pred_delta.loc[aligned_index])
    ).mean()

    # Считаем среднюю signed error для оценки систематического смещения
    mean_signed_error = float((y_pred - y_true).mean())

    # Возвращаем словарь с метриками
    return {
        "directional_accuracy": float(directional_accuracy),
        "mean_signed_error": mean_signed_error,
    }


# Компактный расчёт всех ключевых регрессионных метрик
def regression_metrics(y_true: pd.Series, y_pred: pd.Series) -> Dict[str, float]:
    """
    Description:
    ---------------
        Собирает единый набор метрик качества для сравнения моделей
        в задаче прогноза производственной концентрации.

    Args:
    ---------------
        y_true: Фактические значения целевой переменной
        y_pred: Предсказанные значения модели

    Returns:
    ---------------
        Словарь с метриками MAE, RMSE, MAPE и WAPE

    Raises:
    ---------------
        ValueError: Если длины y_true и y_pred не совпадают

    Examples:
    ---------------
        >>> regression_metrics(
        ...     pd.Series([1.0, 2.0]),
        ...     pd.Series([1.2, 1.8])
        ... )['mae']
        0.2
    """
    # Проверяем согласованность размеров входных рядов
    if len(y_true) != len(y_pred):
        raise ValueError("Размеры y_true и y_pred должны совпадать.")

    # Собираем словарь с регрессионными метриками
    metrics = {
        "mae": float(mean_absolute_error(y_true, y_pred)),
        "rmse": float(np.sqrt(mean_squared_error(y_true, y_pred))),
        "mape": float(safe_mape(y_true, y_pred)),
        "wape": float(safe_wape(y_true, y_pred)),
    }

    # Добавляем метрики направления изменения процесса
    metrics.update(compute_directional_metrics(y_true=y_true, y_pred=y_pred))

    # Возвращаем единый словарь метрик
    return metrics
```

```python
# Функция централизованной загрузки телеметрии и лабораторных замеров
def load_datasets(data_dir: Path) -> Dict[str, pd.DataFrame]:
    """
    Description:
    ---------------
        Загружает обучающие и тестовые таблицы, приводит временные
        колонки к datetime и сортирует наблюдения по времени.

    Args:
    ---------------
        data_dir: Каталог, в котором лежат CSV-файлы проекта

    Returns:
    ---------------
        Словарь с четырьмя датафреймами: train/test telemetry и target

    Raises:
    ---------------
        FileNotFoundError: Если один из обязательных CSV отсутствует
        ValueError: Если после парсинга времени в колонках появились NaT

    Examples:
    ---------------
        >>> datasets = load_datasets(Path("data"))
        >>> sorted(datasets.keys())
        ['data_test', 'data_train', 'target_test', 'target_train']
    """
    # Формируем словарь с обязательными путями к данным
    file_map = {
        "data_train":   data_dir / "data_train.csv",
        "target_train": data_dir / "target_train.csv",
        "data_test":    data_dir / "data_test_small.csv",
        "target_test":  data_dir / "target_test_small.csv",
    }

    # Проверяем наличие каждого файла на диске
    for dataset_name, dataset_path in file_map.items():
        if not dataset_path.exists():
            raise FileNotFoundError(f"Не найден файл {dataset_name}: {dataset_path}")

    # Загружаем таблицы в память с последующей обработкой времени
    data_train   = pd.read_csv(file_map["data_train"])
    target_train = pd.read_csv(file_map["target_train"])
    data_test    = pd.read_csv(file_map["data_test"])
    target_test  = pd.read_csv(file_map["target_test"])

    # Приводим временные колонки телеметрии к datetime
    data_train["datetime"] = pd.to_datetime(data_train["datetime"])
    data_test["datetime"]  = pd.to_datetime(data_test["datetime"])

    # Приводим временные колонки лаборатории к datetime и единому имени
    target_train = target_train.rename(columns={"Дата": "datetime"})
    target_test  = target_test.rename(columns={"Дата": "datetime"})
    target_train["datetime"] = pd.to_datetime(target_train["datetime"])
    target_test["datetime"]  = pd.to_datetime(target_test["datetime"])

    # Проверяем корректность парсинга временных меток
    for frame_name, frame in {
        "data_train":   data_train,
        "target_train": target_train,
        "data_test":    data_test,
        "target_test":  target_test,
    }.items():
        if frame["datetime"].isna().any():
            raise ValueError(
                f"В таблице {frame_name} есть некорректные datetime-значения."
            )

    # Удаляем дубликаты времени через агрегацию среднего по минуте
    data_train = (
        data_train.groupby("datetime", as_index=False)
        .mean(numeric_only=True)
        .sort_values("datetime")
    )
    data_test = (
        data_test.groupby("datetime", as_index=False)
        .mean(numeric_only=True)
        .sort_values("datetime")
    )
    target_train = target_train.sort_values("datetime").drop_duplicates("datetime")
    target_test  = target_test.sort_values("datetime").drop_duplicates("datetime")

    # Возвращаем подготовленные датафреймы в словаре
    return {
        "data_train":   data_train,
        "target_train": target_train,
        "data_test":    data_test,
        "target_test":  target_test,
    }


# Загружаем все таблицы проекта в единый объект
frames = load_datasets(data_dir=DATA_DIR)

# Извлекаем отдельные датафреймы для удобства чтения ноутбука
telemetry_train = frames["data_train"]
target_train    = frames["target_train"]
telemetry_test  = frames["data_test"]
target_test     = frames["target_test"]

# Получаем список всех телеметрических признаков
telemetry_columns = [
    column_name
    for column_name in telemetry_train.columns
    if column_name != "datetime"
]

# Собираем краткую витрину по формам таблиц
shape_summary = pd.DataFrame(
    {
        "dataset": [
            "telemetry_train",
            "target_train",
            "telemetry_test",
            "target_test",
        ],
        "rows": [
            len(telemetry_train),
            len(target_train),
            len(telemetry_test),
            len(target_test),
        ],
        "columns": [
            telemetry_train.shape[1],
            target_train.shape[1],
            telemetry_test.shape[1],
            target_test.shape[1],
        ],
        "start": [
            telemetry_train["datetime"].min(),
            target_train["datetime"].min(),
            telemetry_test["datetime"].min(),
            target_test["datetime"].min(),
        ],
        "end": [
            telemetry_train["datetime"].max(),
            target_train["datetime"].max(),
            telemetry_test["datetime"].max(),
            target_test["datetime"].max(),
        ],
    }
)

# Показываем сводную информацию по размерам и временным диапазонам
shape_summary
```

```python
# Функция для построения витрины по пропускам и временной частоте
def summarize_data_quality(
    telemetry_frame: pd.DataFrame,
    target_frame: pd.DataFrame,
) -> Dict[str, pd.DataFrame]:
    """
    Description:
    ---------------
        Формирует набор диагностических таблиц для EDA: долю пропусков,
        распределение временных шагов и базовую статистику по целевой
        переменной.

    Args:
    ---------------
        telemetry_frame: Таблица минутной телеметрии
        target_frame: Таблица лабораторных измерений цели

    Returns:
    ---------------
        Словарь с датафреймами по пропускам, шагам и статистикам

    Raises:
    ---------------
        ValueError: Если входные таблицы не содержат колонку datetime

    Examples:
    ---------------
        >>> summary = summarize_data_quality(telemetry_train, target_train)
        >>> 'missing_share' in summary
        True
    """
    # Проверяем наличие временной колонки в телеметрии
    if "datetime" not in telemetry_frame.columns:
        raise ValueError("В telemetry_frame отсутствует колонка datetime.")

    # Проверяем наличие временной колонки в таргете
    if "datetime" not in target_frame.columns:
        raise ValueError("В target_frame отсутствует колонка datetime.")

    # Считаем долю пропусков по всем телеметрическим колонкам
    missing_share = (
        telemetry_frame.isna().mean().sort_values(ascending=False).rename("missing_share")
    )

    # Считаем распределение шага во времени для телеметрии
    telemetry_step = (
        telemetry_frame["datetime"].diff().dropna().dt.total_seconds().div(60)
    )

    # Считаем распределение шага во времени для лабораторных измерений
    target_step = target_frame["datetime"].diff().dropna().dt.total_seconds().div(60)

    # Формируем статистики по лабораторной цели
    target_stats = target_frame["target"].describe().to_frame(name="target")

    # Возвращаем набор диагностических таблиц
    return {
        "missing_share": missing_share.to_frame(),
        "telemetry_step": telemetry_step.describe().to_frame(name="minutes"),
        "target_step": target_step.describe().to_frame(name="minutes"),
        "target_stats": target_stats,
    }


# Строим витрину качества данных на обучающем наборе
quality_summary = summarize_data_quality(
    telemetry_frame=telemetry_train,
    target_frame=target_train,
)

# Показываем таблицу с долями пропусков по телеметрии
quality_summary["missing_share"].head(10)

# Создаём холст для набора ключевых графиков EDA
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(20, 12))

# Выбираем четыре первых телеметрических сигнала для обзорного тренда
sample_columns = telemetry_columns[:4]

# Отрисовываем тренды выбранных телеметрических рядов на подвыборке
for axis, column_name in zip(axes[0], sample_columns[:2]):
    # Берём окно начала ряда, чтобы график оставался читаемым
    telemetry_train.head(3000).plot(
        x="datetime",
        y=column_name,
        ax=axis,
        linewidth=1.2,
        color="#667eea",
    )

    # Добавляем заголовок конкретного ряда
    axis.set_title(f"Первые 3000 минут: {column_name}")

    # Убираем подпись оси X для компактности
    axis.set_xlabel("")

# Строим гистограмму распределения целевой переменной
sns.histplot(
    target_train["target"],
    bins=30,
    kde=True,
    ax=axes[1, 0],
    color="#764ba2",
)

# Подписываем график распределения цели
axes[1, 0].set_title("Распределение лабораторной цели")

# Строим гистограмму интервалов между лабораторными замерами
target_gap_minutes = (
    target_train["datetime"].diff().dropna().dt.total_seconds().div(60)
)

# Рисуем распределение интервалов между пробами
sns.histplot(
    target_gap_minutes,
    bins=30,
    ax=axes[1, 1],
    color="#48bb78",
)

# Подписываем график временных интервалов лаборатории
axes[1, 1].set_title("Интервалы между лабораторными пробами, мин")

# Добавляем плотный layout для лучшей читаемости
plt.tight_layout()

# Выводим диагностические таблицы по временным шагам и цели
quality_summary["telemetry_step"], quality_summary["target_step"], quality_summary["target_stats"]
```

```python
# ─── Расширенный EDA: все 16 сигналов телеметрии ────────────────────────────

# Агрегируем минутную телеметрию по часу для визуализации 8-месячных трендов
n_sig = len(telemetry_columns)
n_cols_g, n_rows_g = 4, (n_sig + 3) // 4

fig_ts, axes_ts = plt.subplots(nrows=n_rows_g, ncols=n_cols_g, figsize=(24, n_rows_g * 3.5))
axes_ts_flat = axes_ts.flatten()

for idx, col_name in enumerate(telemetry_columns):
    ax = axes_ts_flat[idx]
    hourly = telemetry_train.set_index("datetime")[col_name].resample("1h").mean()
    ax.plot(hourly.index, hourly.values, linewidth=0.7, color="#667eea")
    # Выделяем выбросы (IQR-правило) красными точками
    q1v, q3v = hourly.quantile(0.25), hourly.quantile(0.75)
    iqrv = q3v - q1v
    out_mask = (hourly < q1v - 1.5 * iqrv) | (hourly > q3v + 1.5 * iqrv)
    if out_mask.sum() > 0:
        ax.scatter(hourly.index[out_mask], hourly[out_mask], color="#e53e3e", s=6, zorder=5)
    ax.set_title(col_name, fontsize=9)
    ax.tick_params(axis="x", rotation=30, labelsize=6)

for idx in range(n_sig, len(axes_ts_flat)):
    axes_ts_flat[idx].set_visible(False)

fig_ts.suptitle(
    "Тренды всех телеметрических сигналов (агрег. по часу; красные точки = выбросы IQR)",
    fontsize=12, y=1.01
)
plt.tight_layout()
plt.show()

# ─── Количественный отчёт по выбросам ────────────────────────────────────────

def detect_outlier_share(telemetry_frame: pd.DataFrame, features: List[str]) -> pd.DataFrame:
    # Рассчитывает долю выбросов методом IQR для каждого сигнала
    rows = []
    for col in features:
        s = telemetry_frame[col].dropna()
        if len(s) == 0:
            continue
        q1s, q3s = s.quantile(0.25), s.quantile(0.75)
        iqrs = q3s - q1s
        flag = (s < q1s - 1.5 * iqrs) | (s > q3s + 1.5 * iqrs)
        rows.append({
            "signal": col,
            "outlier_%": round(flag.mean() * 100, 2),
            "n_outliers": int(flag.sum()),
            "lower_bound": round(q1s - 1.5 * iqrs, 3),
            "upper_bound": round(q3s + 1.5 * iqrs, 3),
        })
    return pd.DataFrame(rows).sort_values("outlier_%", ascending=False).reset_index(drop=True)


outlier_report = detect_outlier_share(telemetry_train, telemetry_columns)

# ─── Тепловая карта пропусков по сигналам и неделям ─────────────────────────

missing_pivot = (
    telemetry_train.set_index("datetime")[telemetry_columns]
    .resample("7D").apply(lambda x: x.isna().mean()).T
)

fig_miss, axes_miss = plt.subplots(1, 2, figsize=(20, 6))

sns.heatmap(
    missing_pivot, ax=axes_miss[0], cmap="YlOrRd",
    cbar_kws={"label": "Доля пропусков"}, linewidths=0.2, linecolor="white",
)
axes_miss[0].set_title("Тепловая карта пропусков по сигналам (7-дневное окно)")
axes_miss[0].set_xlabel("Неделя")
axes_miss[0].set_ylabel("Сигнал")
axes_miss[0].tick_params(axis="x", rotation=45, labelsize=7)

sns.barplot(data=outlier_report, y="signal", x="outlier_%", palette="Reds_r", ax=axes_miss[1])
axes_miss[1].set_title("Доля выбросов по методу IQR (1.5 * IQR), %")
axes_miss[1].set_xlabel("% наблюдений")
axes_miss[1].set_ylabel("")

plt.tight_layout()
plt.show()

print("Топ-5 сигналов по доле выбросов:")
outlier_report.head()
```

  <span style="background: linear-gradient(135deg, #667eea, #764ba2); -webkit-background-clip: text; -webkit-text-fill-color: transparent; margin-right: 8px;">🔄</span> Синхронизация источников и лаги
</h2>

Это центральный этап всей работы. Мы не знаем наверняка, в какой именно момент времени лабораторный результат должен сопоставляться с телеметрией: в момент получения результата, в момент отбора пробы или в окрестности сдвига. Поэтому задержку лучше не «угадать», а проверить по нескольким кандидатам.

<div style="background: #faf5ff; border-left: 4px solid #b794f4; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #2d3748;">
  <div style="font-weight: 700; color: #6b46c1; margin-bottom: 12px; font-size: 0.95em;">📐 Ключевая формула</div>
  <div style="line-height: 1.7;">
    <p style="margin: 0 0 12px 0; font-style: italic; font-size: 1.1em; text-align: center; background: #fff; padding: 10px; border-radius: 6px; border: 1px solid #e9d8fd;">
      Для кандидата задержки <em>d</em> минут строим эффективное время отбора:<br>
      t<sub>sample</sub> = t<sub>lab</sub> - d
    </p>
    <div style="font-size: 0.95em;">
      Затем сопоставляем лабораторный таргет с ближайшим состоянием телеметрии в момент <em>t<sub>sample</sub></em>.
      <br><br>
      Где:
      <ul style="list-style: none; padding: 0; margin: 8px 0 0 0; line-height: 1.6;">
        <li style="margin-bottom: 4px;">
          • <strong>t<sub>lab</sub></strong> — timestamp лабораторного результата в таблице целей;
        </li>
        <li style="margin-bottom: 4px;">
          • <strong>d</strong> — предполагаемая задержка между отбором и получением результата;
        </li>
        <li>
          • <strong>t<sub>sample</sub></strong> — восстановленный момент, когда проба лучше всего описывает физическое состояние печи.
        </li>
      </ul>
    </div>
  </div>
</div>

<div style="font-family: sans-serif;">

  <!-- Шаг 1 -->
  <div style="display: flex; align-items: flex-start; margin-bottom: 20px;">

      Проверяем все значения из диапазона 10–15 минут, согласованные с процессом лаборатории.

      Выбираем задержку, при которой средняя абсолютная корреляция между телеметрией и целью наиболее велика.

```python
# Функция выбора лучшей задержки по корреляционному критерию
def select_best_delay(
    telemetry_frame: pd.DataFrame,
    target_frame: pd.DataFrame,
    delay_candidates: Iterable[int],
    telemetry_features: List[str],
) -> Tuple[int, pd.DataFrame]:
    """
    Description:
    ---------------
        Проверяет несколько допустимых задержек лаборатории и выбирает
        ту, при которой телеметрия лучше согласуется с лабораторной
        целью по средней абсолютной корреляции.

    Args:
    ---------------
        telemetry_frame: Таблица минутной телеметрии
        target_frame: Таблица лабораторной цели
        delay_candidates: Набор проверяемых задержек в минутах
        telemetry_features: Список телеметрических колонок

    Returns:
    ---------------
        Кортеж из лучшей задержки и таблицы результатов перебора

    Raises:
    ---------------
        ValueError: Если после merge_asof не удаётся получить наблюдения

    Examples:
    ---------------
        >>> delay, report = select_best_delay(
        ...     telemetry_train,
        ...     target_train,
        ...     range(10, 16),
        ...     telemetry_columns,
        ... )
        >>> delay in range(10, 16)
        True
    """
    # Создаём контейнер для результатов по каждой задержке
    candidate_rows: List[Dict[str, Any]] = []

    # Перебираем каждую допустимую задержку лаборатории
    for delay_minutes in delay_candidates:
        # Копируем таблицу цели, чтобы безопасно сдвинуть время отбора
        shifted_target = target_frame.copy()

        # Восстанавливаем предполагаемый момент физического отбора пробы
        shifted_target["sample_time"] = (
            shifted_target["datetime"] - pd.Timedelta(minutes=delay_minutes)
        )

        # Присоединяем ближайшую прошлую телеметрию к лабораторному замеру
        merged_frame = pd.merge_asof(
            shifted_target.sort_values("sample_time"),
            telemetry_frame.sort_values("datetime"),
            left_on="sample_time",
            right_on="datetime",
            direction="backward",
            tolerance=pd.Timedelta(minutes=10),
        )

        # Оставляем строки, где лабораторная цель действительно известна
        merged_frame = merged_frame.dropna(subset=["target"]).copy()

        # Считаем, сколько телеметрических сигналов доступно в каждой строке
        available_feature_count = merged_frame[telemetry_features].notna().sum(axis=1)

        # Удаляем строки, где отсутствуют вообще все телеметрические сигналы
        merged_frame = merged_frame.loc[available_feature_count > 0].copy()

        # Проверяем, что синхронизированные данные вообще получены
        if merged_frame.empty:
            continue

        # Считаем pairwise-корреляцию каждого признака с лабораторной целью
        correlation_series = merged_frame[telemetry_features].corrwith(
            merged_frame["target"]
        )

        # Удаляем признаки, для которых корреляция не определилась
        correlation_series = correlation_series.dropna()

        # Если после фильтрации не осталось ни одного признака, пропускаем лаг
        if correlation_series.empty:
            continue

        # Усредняем абсолютную силу связи только по доступным сигналам
        correlation_strength = correlation_series.abs().mean()

        # Сохраняем результат текущей задержки в отчёт
        candidate_rows.append(
            {
                "delay_minutes": delay_minutes,
                "samples": len(merged_frame),
                "mean_abs_corr": float(correlation_strength),
            }
        )

    # Проверяем, что хотя бы один кандидат задержки был корректно оценён
    if not candidate_rows:
        raise ValueError(
            "Не удалось оценить задержку: после синхронизации не осталось "
            "признаков с валидной корреляцией."
        )

    # Формируем таблицу отчёта по всем кандидатам задержки
    delay_report = pd.DataFrame(candidate_rows).sort_values(
        by="mean_abs_corr",
        ascending=False,
    )

    # Извлекаем лучшую задержку из первой строки отчёта
    best_delay = int(delay_report.iloc[0]["delay_minutes"])

    # Возвращаем найденную задержку и детальный отчёт
    return best_delay, delay_report


# Функция синхронизации target с телеметрией после выбора задержки
def align_target_with_telemetry(
    telemetry_frame: pd.DataFrame,
    target_frame: pd.DataFrame,
    delay_minutes: int,
) -> pd.DataFrame:
    """
    Description:
    ---------------
        Сдвигает лабораторный таргет на величину задержки и присоединяет
        к нему телеметрические признаки из ближайшей предыдущей минуты.

    Args:
    ---------------
        telemetry_frame: Таблица минутной телеметрии
        target_frame: Таблица лабораторной цели
        delay_minutes: Выбранная задержка лаборатории в минутах

    Returns:
    ---------------
        Синхронизированный датафрейм с колонкой sample_time

    Raises:
    ---------------
        ValueError: Если после merge_asof не осталось данных

    Examples:
    ---------------
        >>> aligned = align_target_with_telemetry(
        ...     telemetry_train,
        ...     target_train,
        ...     10,
        ... )
        >>> 'sample_time' in aligned.columns
        True
    """
    # Копируем таблицу цели, чтобы добавить технические поля
    aligned_target = target_frame.copy()

    # Вычисляем момент отбора пробы, который предшествует результату
    aligned_target["sample_time"] = (
        aligned_target["datetime"] - pd.Timedelta(minutes=delay_minutes)
    )

    # Выполняем асинхронное объединение с телеметрией по ближайшему прошлому часу
    aligned_frame = pd.merge_asof(
        aligned_target.sort_values("sample_time"),
        telemetry_frame.sort_values("datetime"),
        left_on="sample_time",
        right_on="datetime",
        direction="backward",
        tolerance=pd.Timedelta(minutes=10),
    )

    # Удаляем строки, где слияние не нашло валидную телеметрию
    aligned_frame = aligned_frame.dropna(subset=["target"])

    # Проверяем, что итоговая выборка не пуста
    if aligned_frame.empty:
        raise ValueError("После синхронизации данных не осталось наблюдений.")

    # Возвращаем синхронизированный набор наблюдений
    return aligned_frame


# Подбираем задержку на обучающем наборе данных
best_delay_minutes, delay_report = select_best_delay(
    telemetry_frame=telemetry_train,
    target_frame=target_train,
    delay_candidates=LAB_DELAY_CANDIDATES,
    telemetry_features=telemetry_columns,
)

# Выводим сравнительный отчёт по задержкам
print(f"Выбранная задержка: {best_delay_minutes} минут")
delay_report
```

<h2 id="section-inzhiniring-priznakov" style="color: #ffffff; font-size: 1.8em; font-weight: 700; border-bottom: 3px solid #667eea; padding-bottom: 12px; margin-top: 40px; margin-bottom: 20px; letter-spacing: -0.01em;">
  <span style="background: linear-gradient(135deg, #667eea, #764ba2); -webkit-background-clip: text; -webkit-text-fill-color: transparent; margin-right: 8px;">🧰</span> Инжиниринг признаков
</h2>

Для инерционного агрегата текущая концентрация определяется не только текущими показаниями датчиков, но и траекторией, по которой система пришла в текущее состояние. Поэтому признаки должны кодировать память процесса.

<div style="background: #faf5ff; border-left: 4px solid #b794f4; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #2d3748;">
  <div style="font-weight: 700; color: #6b46c1; margin-bottom: 12px; font-size: 0.95em;">📐 Ключевая формула</div>
  
  <div style="background: white; padding: 15px; border-radius: 6px; border: 1px solid #e9d8fd; margin-bottom: 15px; text-align: center; font-style: italic;">
    Для каждого сигнала x<sub>t</sub> используем лаги и скользящие статистики:<br><br>
    <strong>x<sub>t</sub><sup>(lag, k)</sup> = x<sub>t-k</sub></strong>
    &nbsp;&nbsp;&nbsp;&nbsp;
    <strong>μ<sub>t</sub><sup>(w)</sup> = (1/w) ∑<sub>i=0</sub><sup>w-1</sup> x<sub>t-i</sub></strong>
    &nbsp;&nbsp;&nbsp;&nbsp;
    <strong>σ<sub>t</sub><sup>(w)</sup> = √[(1/w) ∑<sub>i=0</sub><sup>w-1</sup> (x<sub>t-i</sub> - μ<sub>t</sub><sup>(w)</sup>)²]</strong>
  </div>
  
  <div style="font-size: 0.95em; line-height: 1.6;">
    <strong>Где:</strong><br>
    • <strong>x<sub>t-k</sub></strong> — значение сигнала k минут назад<br>
    • <strong>μ<sub>t</sub><sup>(w)</sup></strong> — среднее по окну длиной w<br>
    • <strong>σ<sub>t</sub><sup>(w)</sup></strong> — локальная вариативность сигнала в том же окне
  </div>
</div>

<div style="background: #1a202c; color: #a0aec0; padding: 8px 16px; border-radius: 10px 10px 0 0; margin-bottom: -15px; font-family: 'Fira Code', monospace; font-size: 0.8em; display: flex; justify-content: space-between; align-items: center;">
  <span>🐍 Построение временных признаков</span>
  <span style="background: #2d3748; padding: 2px 10px; border-radius: 10px; font-size: 0.85em;">Python</span>
</div>

```python
# Функция инженерии признаков на минутной телеметрии
def build_time_features(
    telemetry_frame: pd.DataFrame,
    telemetry_features: List[str],
    lag_steps: List[int],
    rolling_windows: List[int],
) -> pd.DataFrame:
    """
    Description:
    ---------------
        Строит лаговые признаки, скользящие статистики, первые разности
        и циклические календарные признаки для минутной телеметрии.

    Args:
    ---------------
        telemetry_frame: Таблица исходной телеметрии
        telemetry_features: Список колонок с числовыми сигналами
        lag_steps: Список лагов в минутах
        rolling_windows: Список окон для rolling-статистик

    Returns:
    ---------------
        Датафрейм с расширенным набором временных признаков

    Raises:
    ---------------
        ValueError: Если датафрейм пуст или не содержит datetime

    Examples:
    ---------------
        >>> feature_frame = build_time_features(
        ...     telemetry_train,
        ...     telemetry_columns,
        ...     [1, 5],
        ...     [5, 15],
        ... )
        >>> 'hour_sin' in feature_frame.columns
        True
    """
    # Проверяем, что датафрейм не пуст
    if telemetry_frame.empty:
        raise ValueError("Нельзя строить признаки на пустом датафрейме.")

    # Проверяем наличие временной колонки
    if "datetime" not in telemetry_frame.columns:
        raise ValueError("В telemetry_frame отсутствует колонка datetime.")

    # Копируем исходную телеметрию для безопасной инженерии
    feature_frame = telemetry_frame.copy().sort_values("datetime")

    # Строим лаговые признаки для каждого телеметрического сигнала
    for column_name in telemetry_features:
        # Перебираем все выбранные лаги для текущего сигнала
        for lag_step in lag_steps:
            # Сдвигаем ряд назад по времени на lag_step минут
            feature_frame[f"{column_name}_lag_{lag_step}"] = (
                feature_frame[column_name].shift(lag_step)
            )

        # Строим скользящие статистики по нескольким окнам
        for window_size in rolling_windows:
            # Создаём rolling-объект для текущего сигнала
            rolling_object = feature_frame[column_name].rolling(window=window_size)

            # Добавляем среднее значение по окну
            feature_frame[f"{column_name}_roll_mean_{window_size}"] = (
                rolling_object.mean()
            )

            # Добавляем стандартное отклонение по окну
            feature_frame[f"{column_name}_roll_std_{window_size}"] = (
                rolling_object.std()
            )

            # Добавляем минимальное значение по окну
            feature_frame[f"{column_name}_roll_min_{window_size}"] = (
                rolling_object.min()
            )

            # Добавляем максимальное значение по окну
            feature_frame[f"{column_name}_roll_max_{window_size}"] = (
                rolling_object.max()
            )

        # Добавляем первую разность как приближение производной
        feature_frame[f"{column_name}_diff_1"] = feature_frame[column_name].diff(1)

        # Добавляем относительное изменение как темп роста сигнала
        feature_frame[f"{column_name}_pct_change_5"] = (
            feature_frame[column_name].pct_change(5, fill_method=None).replace([np.inf, -np.inf], np.nan)
        )

    # Извлекаем час суток для циклического кодирования времени
    feature_frame["hour"] = feature_frame["datetime"].dt.hour

    # Извлекаем день недели для учёта смен и недельных циклов
    feature_frame["day_of_week"] = feature_frame["datetime"].dt.dayofweek

    # Кодируем час суток через синус для циклической природы времени
    feature_frame["hour_sin"] = np.sin(2 * np.pi * feature_frame["hour"] / 24)

    # Кодируем час суток через косинус для сохранения фазы цикла
    feature_frame["hour_cos"] = np.cos(2 * np.pi * feature_frame["hour"] / 24)

    # Кодируем день недели через синус
    feature_frame["weekday_sin"] = np.sin(
        2 * np.pi * feature_frame["day_of_week"] / 7
    )

    # Кодируем день недели через косинус
    feature_frame["weekday_cos"] = np.cos(
        2 * np.pi * feature_frame["day_of_week"] / 7
    )

    # Возвращаем полный набор инженерных признаков
    return feature_frame


# Строим расширенные признаки для train и test телеметрии
telemetry_train_features = build_time_features(
    telemetry_frame=telemetry_train,
    telemetry_features=telemetry_columns,
    lag_steps=LAG_STEPS,
    rolling_windows=ROLLING_WINDOWS,
)
telemetry_test_features = build_time_features(
    telemetry_frame=telemetry_test,
    telemetry_features=telemetry_columns,
    lag_steps=LAG_STEPS,
    rolling_windows=ROLLING_WINDOWS,
)

# Синхронизируем лабораторную цель с расширенной train-телеметрией
aligned_train = align_target_with_telemetry(
    telemetry_frame=telemetry_train_features,
    target_frame=target_train,
    delay_minutes=best_delay_minutes,
)

# Синхронизируем лабораторную цель с расширенной test-телеметрией
aligned_test = align_target_with_telemetry(
    telemetry_frame=telemetry_test_features,
    target_frame=target_test,
    delay_minutes=best_delay_minutes,
)

# Удаляем технические колонки и отделяем target
excluded_columns = ["datetime_x", "datetime_y", "datetime", "sample_time", "target"]
feature_columns = [
    column_name
    for column_name in aligned_train.columns
    if column_name not in excluded_columns
]

# Формируем финальные обучающую и тестовую матрицы
X_train = aligned_train[feature_columns].copy()
y_train = aligned_train["target"]
X_test  = aligned_test[feature_columns].copy()
y_test  = aligned_test["target"]

# Находим признаки, которые полностью пусты даже после синхронизации
all_null_feature_names = [
    column_name for column_name in feature_columns if X_train[column_name].notna().sum() == 0
]

# Удаляем полностью пустые признаки из train и test, чтобы не засорять модели
if all_null_feature_names:
    X_train = X_train.drop(columns=all_null_feature_names)
    X_test = X_test.drop(columns=all_null_feature_names)

# Показываем размеры итоговой supervised-выборки
pd.DataFrame(
    {
        "set": ["train", "test"],
        "rows": [len(X_train), len(X_test)],
        "features": [X_train.shape[1], X_test.shape[1]],
    }
)
```

```python
# ─── Анализ размерности признакового пространства ────────────────────────────

print(f"Исходных признаков: {X_train.shape[1]}")
print(f"Обучающих наблюдений: {X_train.shape[0]}")
print(f"Соотношение N/p: {X_train.shape[0] / X_train.shape[1]:.1f}")

# ─── 1. Признаки с низкой дисперсией ─────────────────────────────────────────

X_train_filled = X_train.fillna(X_train.median(numeric_only=True))

feature_std = X_train_filled.std()
feature_mean_abs = X_train_filled.abs().mean().replace(0, np.nan)
feature_cv = (feature_std / feature_mean_abs).fillna(0)

low_var_thresh = 0.001
low_var_feats = feature_cv[feature_cv < low_var_thresh].index.tolist()
print(f"\nПризнаков с CV < {low_var_thresh} (квази-константы): {len(low_var_feats)}")

# ─── 2. Мультиколлинеарность топ-признаков ───────────────────────────────────

# Топ-40 признаков по корреляции с целью для анализа взаимной коллинеарности
top40 = (
    X_train_filled.corrwith(y_train).abs()
    .sort_values(ascending=False).head(40).index.tolist()
)
corr_feat_matrix = X_train_filled[top40].corr().abs()

fig_mc, ax_mc = plt.subplots(figsize=(16, 14))
mask_tri = np.triu(np.ones_like(corr_feat_matrix, dtype=bool))
sns.heatmap(
    corr_feat_matrix, mask=mask_tri, cmap="Blues", ax=ax_mc,
    linewidths=0.1, cbar_kws={"label": "|Корреляция Пирсона|"}, vmin=0, vmax=1,
)
ax_mc.set_title("Мультиколлинеарность: |корреляция| между топ-40 признаками", fontsize=11)
ax_mc.tick_params(axis="x", rotation=90, labelsize=5)
ax_mc.tick_params(axis="y", labelsize=5)
plt.tight_layout()
plt.show()

high_corr_thresh = 0.95
hc_mask = corr_feat_matrix > high_corr_thresh
np.fill_diagonal(hc_mask.values, False)
n_hc_pairs = int(hc_mask.values.sum() // 2)
print(f"Пар признаков с |корреляцией| > {high_corr_thresh}: {n_hc_pairs}")

# ─── 3. PCA: структура признакового пространства ─────────────────────────────

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler as _SS_pca

_scaler_pca = _SS_pca()
X_pca_input = _scaler_pca.fit_transform(X_train_filled.fillna(0))
n_comp_pca = min(60, X_pca_input.shape[1])
pca_mdl = PCA(n_components=n_comp_pca, random_state=42)
pca_mdl.fit(X_pca_input)

cum_var = np.cumsum(pca_mdl.explained_variance_ratio_) * 100
n_comp_90 = int(np.searchsorted(cum_var, 90)) + 1
n_comp_95 = int(np.searchsorted(cum_var, 95)) + 1

fig_pca, axes_pca = plt.subplots(1, 2, figsize=(16, 5))

axes_pca[0].bar(
    range(1, n_comp_pca + 1),
    pca_mdl.explained_variance_ratio_ * 100,
    color="#667eea", width=0.8,
)
axes_pca[0].set_title("Доля объяснённой дисперсии по компонентам PCA")
axes_pca[0].set_xlabel("Компонента")
axes_pca[0].set_ylabel("Дисперсия, %")
axes_pca[0].grid(True, alpha=0.3)

axes_pca[1].plot(
    range(1, n_comp_pca + 1), cum_var,
    "o-", color="#48bb78", linewidth=1.5, markersize=3,
)
axes_pca[1].axhline(90, linestyle="--", color="#e53e3e",
                     label=f"90% -> {n_comp_90} компонент")
axes_pca[1].axhline(95, linestyle="--", color="#f6ad55",
                     label=f"95% -> {n_comp_95} компонент")
axes_pca[1].axvline(n_comp_90, linestyle=":", color="#e53e3e", alpha=0.6)
axes_pca[1].axvline(n_comp_95, linestyle=":", color="#f6ad55", alpha=0.6)
axes_pca[1].set_title(
    f"Кумулятивная объяснённая дисперсия PCA\n"
    f"90% -> {n_comp_90} компонент | 95% -> {n_comp_95} компонент"
)
axes_pca[1].set_xlabel("Число компонент")
axes_pca[1].set_ylabel("Кумулятивная дисперсия, %")
axes_pca[1].legend(fontsize=9)
axes_pca[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nИтог анализа размерности:")
print(f"  Всего признаков: {X_train.shape[1]}")
print(f"  Квази-констант (CV < {low_var_thresh}): {len(low_var_feats)}")
print(f"  Высококоллинеарных пар (|r| > {high_corr_thresh}): {n_hc_pairs}")
print(f"  PCA-компонент для 90% дисперсии: {n_comp_90}")
print(f"  PCA-компонент для 95% дисперсии: {n_comp_95}")
print(f"\nСтратегия редукции:")
print(f"  OLS: ограничен {OLS_FEATURE_LIMIT} признаками (topK по |корреляции с целью|)")
print(f"  Ridge: L2-регуляризация подавляет избыточные признаки автоматически")
print(f"  RF / GBM: tree-based методы нативно справляются с высокой размерностью")
print(f"  PCA опционально: снижает до {n_comp_90} компонент при сохранении 90% дисперсии")
```

```python
# Функция выбора наиболее информативных признаков для OLS
def select_ols_features(
    X_frame: pd.DataFrame,
    y_series: pd.Series,
    max_features: int,
) -> List[str]:
    """
    Description:
    ---------------
        Выбирает ограниченный набор признаков для OLS по абсолютной
        корреляции с целевой переменной, чтобы сохранить устойчивость
        и интерпретируемость статистической модели.

    Args:
    ---------------
        X_frame: Полная матрица признаков
        y_series: Целевая переменная
        max_features: Максимальное число признаков для OLS

    Returns:
    ---------------
        Список названий отобранных признаков

    Raises:
    ---------------
        ValueError: Если после фильтрации не осталось ни одного признака

    Examples:
    ---------------
        >>> selected = select_ols_features(X_train, y_train, 10)
        >>> len(selected) <= 10
        True
    """
    # Считаем корреляцию каждого признака с целью
    correlation_series = X_frame.assign(target=y_series).corr(numeric_only=True)["target"]

    # Отбрасываем саму целевую переменную и сортируем по силе связи
    selected_features = (
        correlation_series.drop(labels=["target"])
        .abs()
        .sort_values(ascending=False)
        .head(max_features)
        .index
        .tolist()
    )

    # Проверяем, что признаки действительно отобраны
    if not selected_features:
        raise ValueError("Не удалось отобрать признаки для OLS-модели.")

    # Возвращаем финальный список признаков
    return selected_features


# Функция обучения всего набора моделей и замера времени обучения
def train_models(
    X_train_frame: pd.DataFrame,
    y_train_series: pd.Series,
    selected_ols_features: List[str],
) -> Dict[str, Any]:
    """
    Description:
    ---------------
        Обучает статистическую OLS-модель и три прикладные ML-модели,
        возвращая обученные объекты и время fit-процедуры.

    Args:
    ---------------
        X_train_frame: Обучающая матрица признаков
        y_train_series: Обучающая целевая переменная
        selected_ols_features: Список признаков для OLS

    Returns:
    ---------------
        Словарь с обученными моделями и метаданными по fit-time

    Raises:
    ---------------
        ValueError: Если обучающая матрица пуста

    Examples:
    ---------------
        >>> selected = select_ols_features(X_train, y_train, 10)
        >>> trained = train_models(X_train, y_train, selected)
        >>> 'ridge' in trained
        True
    """
    # Проверяем наличие строк в обучающей матрице
    if X_train_frame.empty:
        raise ValueError("Нельзя обучать модели на пустом X_train_frame.")

    # Создаём контейнер для обученных моделей и таймингов
    trained_models: Dict[str, Any] = {}

    # Заполняем пропуски медианой перед построением OLS
    X_train_ols = X_train_frame[selected_ols_features].copy()
    X_train_ols = X_train_ols.fillna(X_train_ols.median(numeric_only=True))

    # Добавляем константу для статистической модели
    X_train_ols_const = add_constant(X_train_ols, has_constant="add")

    # Засекаем время обучения OLS
    start_time = perf_counter()

    # Обучаем интерпретируемую статистическую модель
    ols_model = sm.OLS(y_train_series, X_train_ols_const).fit()

    # Сохраняем затраченное время обучения OLS
    trained_models["ols_fit_time_sec"] = perf_counter() - start_time

    # Сохраняем саму OLS-модель
    trained_models["ols"] = ols_model

    # Создаём конвейер для Ridge-регрессии
    ridge_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", Ridge(alpha=1.0)),
        ]
    )

    # Засекаем время обучения Ridge
    start_time = perf_counter()

    # Обучаем Ridge-модель на всех признаках
    ridge_pipeline.fit(X_train_frame, y_train_series)

    # Сохраняем модель и её fit-time
    trained_models["ridge"] = ridge_pipeline
    trained_models["ridge_fit_time_sec"] = perf_counter() - start_time

    # Создаём конвейер для Random Forest
    random_forest_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            (
                "model",
                RandomForestRegressor(
                    n_estimators=300,
                    max_depth=10,
                    min_samples_leaf=5,
                    random_state=42,
                    n_jobs=-1,
                ),
            ),
        ]
    )

    # Засекаем время обучения Random Forest
    start_time = perf_counter()

    # Обучаем Random Forest на train-наборе
    random_forest_pipeline.fit(X_train_frame, y_train_series)

    # Сохраняем модель и время обучения
    trained_models["random_forest"] = random_forest_pipeline
    trained_models["random_forest_fit_time_sec"] = perf_counter() - start_time

    # Создаём конвейер для HistGradientBoosting
    gradient_boosting_pipeline = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            (
                "model",
                HistGradientBoostingRegressor(
                    learning_rate=0.05,
                    max_depth=6,
                    max_iter=300,
                    min_samples_leaf=10,
                    random_state=42,
                ),
            ),
        ]
    )

    # Засекаем время обучения градиентного бустинга
    start_time = perf_counter()

    # Обучаем бустинг на train-наборе
    gradient_boosting_pipeline.fit(X_train_frame, y_train_series)

    # Сохраняем модель и время обучения
    trained_models["hist_gb"] = gradient_boosting_pipeline
    trained_models["hist_gb_fit_time_sec"] = perf_counter() - start_time

    # Возвращаем полный словарь обученных моделей
    return trained_models


# Выбираем подмножество признаков для статистической модели OLS
ols_feature_names = select_ols_features(
    X_frame=X_train,
    y_series=y_train,
    max_features=OLS_FEATURE_LIMIT,
)

# Обучаем весь набор моделей на train-выборке
trained_models = train_models(
    X_train_frame=X_train,
    y_train_series=y_train,
    selected_ols_features=ols_feature_names,
)

# Выводим краткую сводку по статистической модели
trained_models["ols"].summary().tables[1]
```

  <span style="background: linear-gradient(135deg, #667eea, #764ba2); -webkit-background-clip: text; -webkit-text-fill-color: transparent; margin-right: 8px;">🔬</span> Оценка качества и диагностика
</h2>

Хорошая промышленная модель должна быть точной, быстро считать прогноз и правильно отражать направление изменения процесса. Поэтому оцениваем не только MAE/RMSE, но и directional accuracy, остатки, автокорреляцию и вычислительные затраты.

<div style="background: #faf5ff; border-left: 4px solid #b794f4; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #2d3748;">
  <div style="font-weight: 700; color: #6b46c1; margin-bottom: 6px; font-size: 0.95em;">📐 Ключевая формула</div>
  <div style="line-height: 1.7;">
    Основные метрики точности:
    $$MAE =

rac{1}{N} \sum_{i=1}^N |y_i - \hat{y}_i|, \qquad
    RMSE = \sqrt{

rac{1}{N} \sum_{i=1}^N (y_i - \hat{y}_i)^2},$$
    $$WAPE =

rac{\sum_{i=1}^N |y_i - \hat{y}_i|}{\sum_{i=1}^N |y_i|} \cdot 100\%.$$
    <br><br>
    Где:
    <br>- $N$ — число наблюдений на тесте;
    <br>- $y_i$ — фактическое лабораторное значение;
    <br>- $\hat{y}_i$ — предсказание модели.
  </div>
</div>

<div style="background: #1a202c; color: #a0aec0; padding: 8px 16px; border-radius: 10px 10px 0 0; margin-bottom: -15px; font-family: 'Fira Code', monospace; font-size: 0.8em; display: flex; justify-content: space-between; align-items: center;">
  <span>🐍 Сравнение моделей, остатки и диагностика</span>
  <span style="background: #2d3748; padding: 2px 10px; border-radius: 10px; font-size: 0.85em;">Python</span>
</div>

```python
# Функция инференса и сбора тестовых метрик по всем моделям
def evaluate_models(
    trained_models_map: Dict[str, Any],
    X_test_frame: pd.DataFrame,
    y_test_series: pd.Series,
    ols_features: List[str],
) -> Tuple[pd.DataFrame, Dict[str, pd.Series]]:
    """
    Description:
    ---------------
        Выполняет предсказание для всех обученных моделей, считает
        регрессионные и directional-метрики, а также сохраняет прогнозы.

    Args:
    ---------------
        trained_models_map: Словарь с обученными моделями
        X_test_frame: Матрица тестовых признаков
        y_test_series: Истинные значения целевой переменной на тесте
        ols_features: Список признаков, использованных в OLS

    Returns:
    ---------------
        Таблица метрик и словарь прогнозов по моделям

    Raises:
    ---------------
        ValueError: Если тестовая матрица признаков пуста

    Examples:
    ---------------
        >>> report, predictions = evaluate_models(
        ...     trained_models,
        ...     X_test,
        ...     y_test,
        ...     ols_feature_names,
        ... )
        >>> 'mae' in report.columns
        True
    """
    # Проверяем непустоту тестовой выборки
    if X_test_frame.empty:
        raise ValueError("Нельзя оценивать модели на пустом X_test_frame.")

    # Создаём контейнеры для отчёта и прогнозов
    evaluation_rows: List[Dict[str, Any]] = []
    prediction_map: Dict[str, pd.Series] = {}

    # Готовим OLS-признаки и заполняем пропуски медианой train-схемы
    X_test_ols = X_test_frame[ols_features].copy()
    X_test_ols = X_test_ols.fillna(X_test_ols.median(numeric_only=True))
    X_test_ols_const = add_constant(X_test_ols, has_constant="add")

    # Делаем предсказание OLS-моделью
    start_time = perf_counter()
    ols_pred = pd.Series(
        trained_models_map["ols"].predict(X_test_ols_const),
        index=y_test_series.index,
        name="ols",
    )
    ols_predict_time = perf_counter() - start_time

    # Сохраняем прогноз OLS
    prediction_map["ols"] = ols_pred

    # Считаем метрики для OLS
    ols_metrics = regression_metrics(y_true=y_test_series, y_pred=ols_pred)
    ols_metrics.update(
        {
            "model": "ols",
            "aic": float(trained_models_map["ols"].aic),
            "bic": float(trained_models_map["ols"].bic),
            "fit_time_sec": float(trained_models_map["ols_fit_time_sec"]),
            "predict_time_sec": float(ols_predict_time),
        }
    )
    evaluation_rows.append(ols_metrics)

    # Перебираем остальные прикладные ML-модели
    for model_name, fit_time_key in [
        ("ridge", "ridge_fit_time_sec"),
        ("random_forest", "random_forest_fit_time_sec"),
        ("hist_gb", "hist_gb_fit_time_sec"),
    ]:
        # Засекаем время инференса конкретной модели
        start_time = perf_counter()

        # Выполняем предсказание текущей моделью
        current_prediction = pd.Series(
            trained_models_map[model_name].predict(X_test_frame),
            index=y_test_series.index,
            name=model_name,
        )

        # Сохраняем время предсказания
        current_predict_time = perf_counter() - start_time

        # Кладём прогноз в словарь для дальнейших графиков
        prediction_map[model_name] = current_prediction

        # Считаем все регрессионные метрики модели
        current_metrics = regression_metrics(
            y_true=y_test_series,
            y_pred=current_prediction,
        )

        # Добавляем к метрикам служебную информацию
        current_metrics.update(
            {
                "model": model_name,
                "aic": np.nan,
                "bic": np.nan,
                "fit_time_sec": float(trained_models_map[fit_time_key]),
                "predict_time_sec": float(current_predict_time),
            }
        )

        # Добавляем строку отчёта в общий список
        evaluation_rows.append(current_metrics)

    # Формируем итоговую таблицу сравнения моделей
    evaluation_report = pd.DataFrame(evaluation_rows).sort_values(by="rmse")

    # Возвращаем таблицу метрик и прогнозы
    return evaluation_report, prediction_map


# Запускаем оценку всех моделей на отдельном тестовом наборе
model_report, prediction_map = evaluate_models(
    trained_models_map=trained_models,
    X_test_frame=X_test,
    y_test_series=y_test,
    ols_features=ols_feature_names,
)

# Показываем упорядоченную таблицу качества моделей — явный display() нужен,
# так как в этой ячейке есть ещё код после данной строки
from IPython.display import display as _display
_display(model_report)

# Определяем лучшую модель по RMSE
best_model_name = str(model_report.iloc[0]["model"])

# Извлекаем прогноз лучшей модели
best_prediction = prediction_map[best_model_name]

# Вычисляем остатки лучшей модели
best_residuals = y_test - best_prediction

# Создаём холст для временного графика и диагностики остатков
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(20, 12))

# Строим динамику факта и прогноза на тестовом наборе
axes[0, 0].plot(aligned_test["datetime_x"], y_test, label="Факт", color="#2d3748")
axes[0, 0].plot(
    aligned_test["datetime_x"],
    best_prediction,
    label=f"Прогноз: {best_model_name}",
    color="#667eea",
)
axes[0, 0].set_title("Прогноз против лабораторных значений")
axes[0, 0].legend()

# Строим распределение остатков лучшей модели
sns.histplot(best_residuals, bins=30, kde=True, color="#764ba2", ax=axes[0, 1])
axes[0, 1].set_title("Распределение остатков")

# Строим график остатков против предсказаний
axes[1, 0].scatter(best_prediction, best_residuals, alpha=0.7, color="#48bb78")
axes[1, 0].axhline(0, linestyle="--", color="#e53e3e")
axes[1, 0].set_title("Остатки vs предсказания")
axes[1, 0].set_xlabel("Предсказание")
axes[1, 0].set_ylabel("Остаток")

# Строим автокорреляцию остатков
plot_acf(best_residuals, ax=axes[1, 1], lags=30)
axes[1, 1].set_title("ACF остатков")

# Укладываем графики компактно
plt.tight_layout()

# Считаем статистические тесты по остаткам лучшей модели
ljung_box_table = acorr_ljungbox(best_residuals, lags=[10], return_df=True)
shapiro_sample = best_residuals.sample(min(len(best_residuals), 500), random_state=42)
shapiro_result = stats.shapiro(shapiro_sample)

# Строим дизайн-матрицу для теста Бройша–Пагана
bp_exog = add_constant(pd.DataFrame({"prediction": best_prediction}))
bp_test = het_breuschpagan(best_residuals, bp_exog)

# Формируем витрину с ключевыми статистическими тестами
residual_tests = pd.DataFrame(
    {
        "test": ["Shapiro-Wilk", "Ljung-Box(10)", "Breusch-Pagan"],
        "statistic": [
            float(shapiro_result.statistic),
            float(ljung_box_table["lb_stat"].iloc[0]),
            float(bp_test[0]),
        ],
        "p_value": [
            float(shapiro_result.pvalue),
            float(ljung_box_table["lb_pvalue"].iloc[0]),
            float(bp_test[1]),
        ],
    }
)

# Выводим результаты тестов остатков
residual_tests
```

```python
# ─── Временной ряд остатков лучшей модели ────────────────────────────────────

# График остатков во времени — ключевой инструмент обнаружения систематического дрейфа
test_dt_col = "datetime_x" if "datetime_x" in aligned_test.columns else "sample_time"
test_dt_vals = (
    aligned_test[test_dt_col]
    if test_dt_col in aligned_test.columns
    else pd.RangeIndex(len(y_test))
)

fig_rt, ax_rt = plt.subplots(figsize=(18, 5))

ax_rt.plot(
    test_dt_vals, best_residuals.values,
    linewidth=0.8, color="#667eea", label="Остаток (факт - прогноз)",
)
ax_rt.axhline(0, linestyle="--", color="#e53e3e", linewidth=1.2)

# Скользящее среднее для выявления тренда смещения
window_ma = max(3, len(best_residuals) // 15)
resid_ma = pd.Series(best_residuals.values).rolling(window=window_ma, center=True, min_periods=1).mean()
ax_rt.plot(
    test_dt_vals, resid_ma.values,
    linewidth=2, color="#f6ad55", label=f"MA({window_ma}) остатков (тренд смещения)",
)
ax_rt.fill_between(test_dt_vals, best_residuals.values, 0, alpha=0.12, color="#667eea")

ax_rt.set_title(
    f"Остатки модели '{best_model_name}' во времени\n"
    "Оранжевая линия = тренд смещения; значительные отклонения от 0 = систематическая ошибка"
)
ax_rt.set_xlabel("Дата / лабораторный замер")
ax_rt.set_ylabel("Остаток")
ax_rt.legend()
plt.tight_layout()
plt.show()

# Сегментный анализ дрейфа остатков по квартилям тестового периода
n_q = len(best_residuals)
q_labels = ["Q1 (начало)", "Q2", "Q3", "Q4 (конец)"]
q_bounds = [0, n_q // 4, n_q // 2, 3 * n_q // 4, n_q]
drift_stats = []
for qi in range(4):
    seg = best_residuals.iloc[q_bounds[qi]:q_bounds[qi + 1]]
    drift_stats.append({
        "период": q_labels[qi],
        "mean_residual": round(float(seg.mean()), 4),
        "std_residual": round(float(seg.std()), 4),
        "n_obs": len(seg),
    })
print("Дрейф остатков по квартилям тестового периода:")
print(pd.DataFrame(drift_stats).to_string(index=False))

# ─── TimeSeriesSplit кросс-валидация ─────────────────────────────────────────

from sklearn.base import clone as _clone

# Используем скользящую кросс-валидацию с сохранением временного порядка
cv_splitter = TimeSeriesSplit(n_splits=5)
cv_results = []

# Pipeline содержит Imputer внутри — передаём сырой X_train
for cv_name, cv_model in [
    ("ridge", trained_models["ridge"]),
    ("random_forest", trained_models["random_forest"]),
    ("hist_gb", trained_models["hist_gb"]),
]:
    fold_maes, fold_rmses = [], []
    for train_idx, val_idx in cv_splitter.split(X_train):
        X_cv_tr = X_train.iloc[train_idx]
        y_cv_tr = y_train.iloc[train_idx]
        X_cv_val = X_train.iloc[val_idx]
        y_cv_val = y_train.iloc[val_idx]

        cv_clone = _clone(cv_model)
        cv_clone.fit(X_cv_tr, y_cv_tr)
        cv_pred = cv_clone.predict(X_cv_val)

        fold_maes.append(float(mean_absolute_error(y_cv_val, cv_pred)))
        fold_rmses.append(float(mean_squared_error(y_cv_val, cv_pred) ** 0.5))

    cv_results.append({
        "model": cv_name,
        "cv_mae_mean": round(float(np.mean(fold_maes)), 4),
        "cv_mae_std":  round(float(np.std(fold_maes)), 4),
        "cv_rmse_mean": round(float(np.mean(fold_rmses)), 4),
        "cv_rmse_std":  round(float(np.std(fold_rmses)), 4),
    })

cv_report = pd.DataFrame(cv_results).sort_values("cv_rmse_mean").reset_index(drop=True)
print("\nTimeSeriesSplit (5 фолдов) кросс-валидация на обучающей выборке:")
cv_report
```

```python
# Функция расчёта permutation importance для лучшей прикладной модели
def compute_feature_importance(
    model_name: str,
    trained_models_map: Dict[str, Any],
    X_test_frame: pd.DataFrame,
    y_test_series: pd.Series,
) -> pd.DataFrame:
    """
    Description:
    ---------------
        Рассчитывает permutation importance для лучшей прикладной модели,
        чтобы оценить вклад признаков в качество прогноза на тесте.

    Args:
    ---------------
        model_name: Имя выбранной модели
        trained_models_map: Словарь с обученными моделями
        X_test_frame: Тестовая матрица признаков
        y_test_series: Истинные тестовые значения цели

    Returns:
    ---------------
        Отсортированная таблица важностей признаков

    Raises:
    ---------------
        ValueError: Если выбранная модель — OLS без sklearn-интерфейса

    Examples:
    ---------------
        >>> importance = compute_feature_importance(
        ...     'ridge',
        ...     trained_models,
        ...     X_test,
        ...     y_test,
        ... )
        >>> 'feature' in importance.columns
        True
    """
    # Запрещаем permutation importance для statsmodels OLS напрямую
    if model_name == "ols":
        raise ValueError(
            "Для OLS используйте коэффициенты модели; permutation применяем к sklearn-моделям."
        )

    # Вычисляем permutation importance на тестовой выборке
    importance_result = permutation_importance(
        estimator=trained_models_map[model_name],
        X=X_test_frame,
        y=y_test_series,
        n_repeats=10,
        random_state=42,
        n_jobs=-1,
    )

    # Формируем таблицу с признаками и средней важностью
    importance_frame = pd.DataFrame(
        {
            "feature": X_test_frame.columns,
            "importance_mean": importance_result.importances_mean,
            "importance_std": importance_result.importances_std,
        }
    ).sort_values(by="importance_mean", ascending=False)

    # Возвращаем отсортированную таблицу важностей
    return importance_frame


# Определяем, можно ли посчитать permutation importance для победителя
if best_model_name == "ols":
    # Формируем таблицу коэффициентов OLS как интерпретацию статистической модели
    interpretation_table = (
        trained_models["ols"].params.drop(labels=["const"], errors="ignore")
        .sort_values(key=np.abs, ascending=False)
        .head(15)
        .rename("coefficient")
        .reset_index()
        .rename(columns={"index": "feature"})
    )
else:
    # Считаем permutation importance для лучшей sklearn-модели
    interpretation_table = compute_feature_importance(
        model_name=best_model_name,
        trained_models_map=trained_models,
        X_test_frame=X_test,
        y_test_series=y_test,
    ).head(15)

# Визуализируем топ признаков лучшей модели
plt.figure(figsize=(12, 8))
sns.barplot(
    data=interpretation_table,
    y="feature",
    x=interpretation_table.columns[1],
    palette="viridis",
)
plt.title(f"Ключевые признаки модели: {best_model_name}")
plt.xlabel("Важность / коэффициент")
plt.ylabel("")
plt.tight_layout()

# Показываем таблицу интерпретации
interpretation_table.head(15)
```

```python
# ─── SHAP-анализ лучшей модели ───────────────────────────────────────────────

if not SHAP_AVAILABLE:
    print("SHAP не установлен. Для запуска: pip install shap")
    print("Используйте permutation importance выше как альтернативу.")
else:
    # Выбираем лучшую tree-based модель (SHAP TreeExplainer требует дерево)
    shap_name = best_model_name if best_model_name in ("random_forest", "hist_gb") else "hist_gb"
    shap_pipeline = trained_models[shap_name]

    # Заполняем пропуски медианой для SHAP
    X_shap_raw = X_test.fillna(X_test.median(numeric_only=True))
    # Ограничиваем выборку для ускорения расчёта SHAP-значений
    sample_n = min(200, len(X_shap_raw))
    X_shap_sample = X_shap_raw.sample(sample_n, random_state=42)

    # Трансформируем данные через шаги Pipeline (кроме финальной модели)
    pipeline_steps = list(shap_pipeline.named_steps.items())
    X_shap_transformed = X_shap_sample.copy()
    for step_name, step_obj in pipeline_steps[:-1]:
        X_shap_transformed = step_obj.transform(X_shap_transformed)
    inner_model = pipeline_steps[-1][1]

    # Восстанавливаем DataFrame с именами признаков
    X_shap_df = pd.DataFrame(
        X_shap_transformed,
        columns=X_shap_sample.columns,
        index=X_shap_sample.index,
    )

    # Инициализируем TreeExplainer и вычисляем SHAP-значения
    shap_explainer = shap.TreeExplainer(inner_model)
    shap_vals = shap_explainer.shap_values(X_shap_df)

    # ─── Summary plot (violin / beeswarm) ────────────────────────────────────
    plt.figure(figsize=(12, 9))
    shap.summary_plot(
        shap_vals, X_shap_df, plot_type="violin", max_display=20, show=False,
    )
    plt.title(
        f"SHAP Summary Plot: модель '{shap_name}'\n"
        "Цвет = значение признака (красный=высокое, синий=низкое); "
        "x-ось = влияние на предсказание",
        fontsize=11,
    )
    plt.tight_layout()
    plt.show()

    # ─── Bar plot средней абсолютной важности ────────────────────────────────
    plt.figure(figsize=(10, 8))
    shap.summary_plot(
        shap_vals, X_shap_df, plot_type="bar", max_display=15, show=False,
    )
    plt.title(f"SHAP Feature Importance (mean|SHAP|): модель '{shap_name}'", fontsize=11)
    plt.tight_layout()
    plt.show()

    # Таблица топ-15 признаков по SHAP
    shap_importance = (
        pd.DataFrame({
            "feature": X_shap_df.columns,
            "mean_abs_shap": np.abs(shap_vals).mean(axis=0),
        })
        .sort_values("mean_abs_shap", ascending=False)
        .head(15)
        .reset_index(drop=True)
    )

    print("Топ-15 признаков по среднему |SHAP|:")
    print(shap_importance.to_string(index=False))

    # ─── Физическая интерпретация ─────────────────────────────────────────────
    print("\nФизическая интерпретация топ-признаков SHAP:")
    print("  _lag_N       : значение сигнала N минут назад — 'память' реактора о прошлых состояниях")
    print("  _roll_mean_W : скользящее среднее за W мин — инерция теплового режима установки")
    print("  _roll_std_W  : нестабильность режима — рост std означает переходный/нестационарный режим")
    print("  _diff_1      : мгновенная скорость изменения — ускорение роста/падения концентрации")
    print("  _pct_change  : темп роста — нормированная динамика относительно предыдущего окна")
    print("  hour_sin/cos : суточная цикличность смен и технологических регламентов производства")
    print("\n  Доминирующий тип признаков указывает на то, что концентрация определяется")
    print("  не мгновенным состоянием, а накопленной историей процесса — что физически")
    print("  соответствует инерции химических реакций и тепловых режимов.")
    # ─── Согласование SHAP и Permutation Importance ───────────────────────────
    # Сравниваем два метода интерпретации и объясняем расхождение рейтингов.
    # ВАЖНО: при наличии высококоллинеарных признаков (147 пар с |r|>0.95)
    # permutation importance недооценивает вклад коллинеарных признаков,
    # т.к. модель компенсирует перестановку через связанный признак.
    # SHAP (TreeExplainer) устойчивее к мультиколлинеарности.

    # Получаем топ-15 признаков по обоим методам
    shap_top_feats = set(shap_importance["feature"].head(15))

    # interpretation_table — результат permutation importance из ячейки выше
    perm_col = "importance_mean" if "importance_mean" in interpretation_table.columns else interpretation_table.columns[1]
    perm_top_feats = set(interpretation_table["feature"].head(15))

    # Признаки, попавшие в топ-15 обоих методов (консенсус)
    consensus_feats = shap_top_feats & perm_top_feats
    only_shap = shap_top_feats - perm_top_feats
    only_perm = perm_top_feats - shap_top_feats

    print("\n─── Согласование методов важности признаков ──────────────────────────")
    print(f"Признаков в топ-15 SHAP:               {len(shap_top_feats)}")
    print(f"Признаков в топ-15 Permutation:        {len(perm_top_feats)}")
    print(f"Совпадают в обоих методах (консенсус): {len(consensus_feats)}")
    print(f"Только в SHAP:                         {len(only_shap)}")
    print(f"Только в Permutation:                  {len(only_perm)}")

    if consensus_feats:
        print("\nКонсенсусные признаки (важны по обоим методам):")
        for feat in sorted(consensus_feats):
            shap_rank = shap_importance[shap_importance["feature"] == feat].index[0] + 1
            perm_rank = (interpretation_table[interpretation_table["feature"] == feat].index[0] + 1
                         if feat in interpretation_table["feature"].values else "—")
            print(f"  {feat}  (SHAP ранг: {shap_rank}, Perm ранг: {perm_rank})")

    print("\nПричина расхождения рейтингов:")
    print("  • 147 пар признаков с |r|>0.95 → permutation importance нестабилен:")
    print("    перестановка признака X компенсируется через коллинеарный Y,")
    print("    поэтому падение метрики минимально, и важность занижается.")
    print("  • SHAP (TreeExplainer) разбивает вклад по узлам дерева —")
    print("    корректнее в условиях мультиколлинеарности.")
    print("  • Рекомендация: SHAP является более надёжной оценкой важности")
    print("    в данном контексте; оба метода согласуются на уровне ТИПОВ признаков:")
    print("    lag > rolling_stats > diff/pct_change.")

```
